#### DATA QUALITY OBSERVABILITY

##### 14.1 DOCUMENTAÇÃO DO DATA QUALITY OBSERVABILITY

###### Objetivo

O **Data Quality Observability** é responsável por monitorar a saúde operacional do Data Quality Framework, utilizando os resultados produzidos pelo **Framework Orchestrator (11)** e, opcionalmente, pelo **Data Quality Report (12)**.

Seu objetivo é identificar falhas, bloqueios, gargalos de desempenho, divergências de execução e indisponibilidade de resultados, contribuindo para a confiabilidade operacional e a observabilidade do framework.

O módulo não executa diretamente os Profiles, não consulta o dataset original e não modifica os dados analisados. Sua responsabilidade permanece concentrada no acompanhamento operacional.

A análise busca responder:

> Quais componentes do Data Quality Framework foram executados, quais apresentaram falhas, quanto tempo consumiram e quais situações exigem atenção operacional?

O Data Quality Observability possui caráter **operacional, diagnóstico e preventivo**. O sucesso técnico de uma execução não representa, necessariamente, a aprovação da qualidade do dataset analisado.

---

###### Exemplos

**Exemplo 1 — Execução concluída sem falhas**

Considere que o Orchestrator tenha planejado três Profiles e registrado três execuções com status `SUCCESS`.

O Data Quality Observability consolida os registros e apresenta indicadores de conclusão operacional.

O resultado permite identificar que os componentes foram concluídos tecnicamente, sem afirmar que o dataset está aprovado.

**Exemplo 2 — Falha e dependência bloqueada**

Considere que determinado Profile apresente uma exceção durante a execução, recebendo o status `FAILED`.

Outro Profile, dependente do primeiro, poderá receber o status `SKIPPED_DEPENDENCY`.

O módulo identifica a falha e o bloqueio, permitindo investigar a sequência operacional e os componentes afetados.

**Exemplo 3 — Execução interrompida**

Considere que um Profile esteja presente no planejamento, mas não possua registro correspondente de execução.

O módulo identifica essa divergência e classifica o componente como `NAO_REGISTRADO`.

Essa condição não deve ser interpretada como sucesso ou falha confirmada, pois indica ausência de informação operacional.

**Exemplo 4 — Duração elevada**

Considere que um Profile tenha duração superior ao limite configurado.

O módulo identifica a ocorrência e gera um alerta para investigação do desempenho.

O limite utilizado representa um critério operacional configurável, não uma regra de qualidade dos dados.

**Exemplo 5 — Indisponibilidade de resultados analíticos**

Considere que o Data Quality Report tenha identificado uma fonte indisponível.

O Data Quality Observability poderá utilizar esse inventário para sinalizar a indisponibilidade dos resultados analíticos.

A ausência de uma fonte não significa que o dataset esteja livre de problemas.

---

###### Principais análises

O Data Quality Observability contempla:

1. **Configuração do monitoramento**
   - Definição das fontes operacionais;
   - Configuração dos modos de leitura;
   - Definição dos limites de processamento;
   - Configuração dos limiares de alerta;
   - Identificação da execução monitorada.

2. **Validação dos contratos**
   - Verificação dos DataFrames esperados;
   - Validação dos campos necessários;
   - Identificação de estruturas incompatíveis;
   - Verificação dos identificadores de execução;
   - Registro das fontes indisponíveis.

3. **Monitoramento da saúde operacional**
   - Identificação de execuções concluídas;
   - Identificação de falhas;
   - Identificação de dependências bloqueadas;
   - Identificação de componentes sem registro;
   - Verificação de divergências entre planejamento e execução.

4. **Monitoramento de desempenho**
   - Análise da duração dos Profiles;
   - Consolidação dos tempos registrados;
   - Identificação de execuções demoradas;
   - Comparação com limites configurados;
   - Identificação de possíveis gargalos operacionais.

5. **Análise de eventos**
   - Identificação de mensagens de erro;
   - Registro de falhas operacionais;
   - Identificação de bloqueios;
   - Organização de inconsistências;
   - Consolidação dos eventos relevantes.

6. **Rastreabilidade da execução**
   - Associação dos Profiles ao `run_id`;
   - Identificação da ordem planejada;
   - Associação dos registros aos componentes;
   - Verificação das dependências;
   - Identificação de divergências operacionais.

7. **Disponibilidade dos resultados analíticos**
   - Consulta opcional ao inventário do módulo 12;
   - Identificação de fontes disponíveis;
   - Identificação de resultados ausentes;
   - Verificação de incompatibilidades;
   - Registro de limitações de cobertura.

8. **Classificação de alertas**
   - Aplicação de regras operacionais explícitas;
   - Identificação de situações críticas;
   - Identificação de situações de atenção;
   - Organização das evidências;
   - Disponibilização de alertas para investigação.

9. **Consolidação executiva**
   - Contagem de componentes planejados;
   - Contagem de componentes executados;
   - Consolidação dos status;
   - Apresentação das durações;
   - Identificação dos principais alertas;
   - Construção de indicadores de saúde operacional.

---

###### Métodos de execução

**Leitura em memória**

O modo `MEMORY` utiliza DataFrames disponíveis na sessão Python atual.

Essa estratégia permite analisar resultados operacionais que estejam acessíveis no mesmo contexto de execução.

**Leitura em Delta**

O modo `DELTA` consulta tabelas previamente persistidas e explicitamente configuradas.

Essa estratégia permite observar resultados de execuções anteriores sem executar novamente os Profiles.

**Leitura automática**

O modo `AUTO` prioriza os resultados disponíveis em memória e utiliza tabelas Delta configuradas quando necessário.

**Validação das fontes**

O módulo verifica a disponibilidade e a compatibilidade dos resultados operacionais antes de realizar as análises.

Fontes ausentes ou incompatíveis são registradas para evitar interpretações incorretas.

**Monitoramento por execução**

Os registros são associados ao identificador `run_id`, permitindo distinguir diferentes execuções do framework.

Quando houver múltiplos identificadores disponíveis, a seleção deve ser explícita para evitar mistura de resultados.

**Geração de alertas**

Os alertas são produzidos a partir de critérios operacionais configurados.

O módulo disponibiliza os diagnósticos em DataFrames, sem realizar notificações externas automaticamente.

**Rastreabilidade operacional**

A rastreabilidade utiliza os identificadores dos Profiles, a ordem planejada e os registros de execução.

Essa abordagem não constitui tracing distribuído completo, pois não implementa spans ou instrumentação entre serviços.

---

###### Tratamento de situações especiais

Para preservar a confiabilidade operacional, o Data Quality Observability considera:

- **Fonte ausente:** registrada no inventário, sem simular resultados;
- **Contrato incompatível:** identificado antes da consolidação;
- **Plano indisponível:** permite observar registros existentes, mas não comprova completude;
- **Registro sem planejamento:** sinalizado como divergência operacional;
- **Profile planejado sem registro:** classificado como `NAO_REGISTRADO`;
- **Status desconhecido:** preservado e sinalizado;
- **Duração negativa:** identificada como informação operacional inválida;
- **Duração nula:** tratada como informação indisponível;
- **Múltiplos `run_id`:** exige seleção explícita;
- **Fonte truncada:** impede conclusões de completude;
- **Resultado analítico ausente:** não representa ausência de problemas no dataset;
- **Falha técnica:** não representa automaticamente reprovação da qualidade dos dados.

A ausência de registros operacionais não deve ser interpretada como execução bem-sucedida.

---

###### Interpretação dos resultados

Os status operacionais representam o comportamento dos componentes durante a execução.

**`SUCCESS`**

Indica que a chamada ao notebook foi concluída tecnicamente.

**`FAILED`**

Indica que ocorreu uma falha capturada durante a execução.

**`SKIPPED_DEPENDENCY`**

Indica que o componente não foi executado devido à falha de uma dependência.

**`NAO_REGISTRADO`**

Indica que o componente estava previsto no planejamento, mas não possui registro operacional correspondente.

**Alertas operacionais**

Representam situações que exigem investigação, como falhas, bloqueios, divergências e duração elevada.

Os alertas não modificam o Data Quality Score e não representam notificações externas.

**Princípio fundamental:**

> Observabilidade operacional não significa avaliação da qualidade dos dados. O sucesso técnico dos componentes e a qualidade do dataset são dimensões distintas.

---

###### Características do Data Quality Observability

O módulo foi desenvolvido considerando os seguintes princípios:

- **Genérico:** não depende de regras específicas de negócio;
- **Reutilizável:** pode monitorar execuções de diferentes datasets;
- **Modular:** mantém o monitoramento separado dos Profiles;
- **Configurável:** permite definir fontes e limiares;
- **Rastreável:** preserva identificadores e registros operacionais;
- **Observável:** organiza métricas, eventos e alertas;
- **Não destrutivo:** não modifica diretamente os dados analisados;
- **Conservador:** não interpreta informações ausentes como sucesso;
- **Escalável:** prioriza resultados operacionais agregados;
- **Extensível:** permite incorporar novas fontes de monitoramento;
- **Orientado à investigação:** facilita a identificação de falhas e gargalos;
- **Compatível com DataOps:** apoia o acompanhamento da confiabilidade operacional.

O módulo utiliza resultados previamente produzidos, evitando reexecutar os Profiles ou recalcular métricas sobre o dataset original.

---

###### DataFrames de resultado

O Data Quality Observability disponibiliza estruturas padronizadas para acompanhamento operacional.

**`dq_obs_sources_df`**

Representa o inventário das fontes utilizadas e seus resultados de validação.

**`dq_obs_execution_df`**

Disponibiliza os registros operacionais normalizados.

**`dq_obs_health_df`**

Consolida o estado dos Profiles planejados e executados.

**`dq_obs_performance_df`**

Apresenta indicadores de duração e desempenho operacional.

**`dq_obs_events_df`**

Organiza eventos relacionados a falhas, bloqueios e inconsistências.

**`dq_obs_trace_df`**

Disponibiliza a sequência planejada e os resultados associados aos componentes.

**`dq_obs_analytical_sources_df`**

Representa a disponibilidade opcional das fontes analíticas identificadas pelo módulo 12.

**`dq_obs_alerts_df`**

Consolida os alertas calculados a partir das regras operacionais.

**`dq_obs_summary_df`**

Apresenta uma visão executiva da execução monitorada.

---

###### Limitações

A implementação atual apresenta algumas limitações que devem ser consideradas:

1. **Ausência de persistência automática:** os resultados em memória não constituem histórico durável.

2. **Dependência das fontes operacionais:** o monitoramento exige registros previamente produzidos.

3. **Isolamento de contexto:** variáveis de notebooks executados separadamente não são compartilhadas automaticamente.

4. **Ausência de tracing distribuído:** o módulo não produz spans ou telemetria entre microsserviços.

5. **Ausência de notificações externas:** os alertas são disponibilizados em DataFrames.

6. **Ausência de baseline histórico:** os limites de desempenho são configurados explicitamente, sem cálculo automático de tendências.

7. **Ausência de inferência de causa raiz:** os registros disponíveis fornecem evidências operacionais, mas não comprovam causalidade.

8. **Cobertura limitada:** fontes ausentes ou truncadas podem impedir conclusões sobre a completude da execução.

9. **Dependência do ambiente:** o processamento depende dos recursos disponíveis no Databricks.

Essas limitações deverão orientar futuras evoluções relacionadas à persistência, monitoramento contínuo, alertas automatizados e integração com práticas de **DataOps**.

---

###### Resultado esperado

Ao final da execução, o Data Quality Observability deverá permitir responder:

1. Qual execução foi monitorada?
2. Quantos Profiles foram planejados?
3. Quantos Profiles possuem registros operacionais?
4. Quais componentes foram concluídos com sucesso?
5. Quais apresentaram falhas?
6. Quais foram bloqueados por dependências?
7. Quais componentes não possuem registros?
8. Qual foi a duração de cada Profile?
9. Quais componentes ultrapassaram os limites configurados?
10. Quais mensagens de erro foram identificadas?
11. Quais resultados analíticos estão indisponíveis?
12. Quais alertas exigem investigação?
13. Quais limitações impedem concluir que a execução foi completa?

Os resultados poderão ser utilizados posteriormente na construção de dashboards operacionais, históricos de execução, indicadores de confiabilidade e mecanismos de monitoramento associados às práticas de **DataOps**.

**Observação:** o Data Quality Observability deve permanecer responsável pelo monitoramento operacional do framework. A avaliação da qualidade do dataset pertence aos Profiles, ao Data Quality Score e às políticas de qualidade estabelecidas.

```python
# ============================================================
# 14.3 CONFIGURAÇÕES
# ============================================================

# O QUE FAZ:
# Define parâmetros sem executar notebooks.

# COMO FAZ:
# Declara fontes e limites explícitos.

# POR QUE É IMPORTANTE:
# Evita reprocessamento e leituras não intencionais.

# PERGUNTA RESPONDIDA:
# "Quais fontes e limiares serão usados?"

DQ_OBS_MODE = "MEMORY"  # MEMORY, DELTA, AUTO
DQ_OBS_RUN_ID = None  # Obrigatório se houver mais de um run_id na fonte
DQ_OBS_TABLES = {"PLAN": None, "EXECUTION": None, "REPORT_SOURCES": None}
DQ_OBS_MAX_ROWS = 10000  # Limite por fonte agregada; acima disso, interrompe a leitura dessa fonte
DQ_OBS_DURATION_WARNING_SECONDS = 900.0  # Limiar operacional configurável
DQ_OBS_DISPLAY_RESULTS = True

if DQ_OBS_MODE not in {"MEMORY", "DELTA", "AUTO"}:
    raise ValueError("DQ_OBS_MODE inválido.")
if not isinstance(DQ_OBS_MAX_ROWS, int) or DQ_OBS_MAX_ROWS < 1:
    raise ValueError("DQ_OBS_MAX_ROWS deve ser inteiro positivo.")
if DQ_OBS_DURATION_WARNING_SECONDS <= 0:
    raise ValueError("DQ_OBS_DURATION_WARNING_SECONDS deve ser positivo.")
DQ_OBS_CREATED_AT_UTC = datetime.now(timezone.utc).isoformat()
```

```python
# ============================================================
# 14.4 CATÁLOGO E CONTRATOS
# ============================================================

# O QUE FAZ:
# Declara entradas e schemas mínimos.

# COMO FAZ:
# Usa contratos reais dos notebooks 11 e 12.

# POR QUE É IMPORTANTE:
# Previne inferências incorretas.

# PERGUNTA RESPONDIDA:
# "Quais dados são necessários?"

DQ_OBS_CATALOG = {
    "PLAN": {"variable": "orchestrator_plan_df", "required": {"run_id", "ordem", "profile", "notebook_path"}},
    "EXECUTION": {"variable": "orchestrator_execution_df", "required": {"run_id", "profile", "status", "started_at_utc", "finished_at_utc", "duration_seconds", "message"}},
    "REPORT_SOURCES": {"variable": "dq_report_sources_df", "required": {"dimensao", "status", "origem"}},
}
DQ_OBS_REQUIRED_SOURCES = {"EXECUTION"}
DQ_OBS_KNOWN_STATUSES = {"SUCCESS", "FAILED", "SKIPPED_DEPENDENCY"}

def dq_obs_frame(rows, schema):
    return spark.createDataFrame(rows, schema=T.StructType([
        T.StructField(name, dtype, True) for name, dtype in schema
    ]))

def dq_obs_show(frame):
    if DQ_OBS_DISPLAY_RESULTS:
        display(frame)
```

```python
# ============================================================
# 14.5 LEITURA E VALIDAÇÃO DAS FONTES
# ============================================================

# O QUE FAZ:
# Consulta somente DataFrames existentes ou tabelas configuradas.

# COMO FAZ:
# Valida contrato, tamanho e seleção de execução.

# POR QUE É IMPORTANTE:
# Evita mistura de runs e conclusões com fontes truncadas.

# PERGUNTA RESPONDIDA:
# "Quais fontes são utilizáveis?"

dq_obs_data = {}
dq_obs_source_records = []
for key, spec in DQ_OBS_CATALOG.items():
    frame, origin, status, detail = None, None, "AUSENTE", "Fonte não configurada ou indisponível"
    if DQ_OBS_MODE in {"MEMORY", "AUTO"}:
        candidate = globals().get(spec["variable"])
        if isinstance(candidate, DataFrame):
            frame, origin, status, detail = candidate, "MEMORY", "DISPONIVEL", ""
    if frame is None and DQ_OBS_MODE in {"DELTA", "AUTO"} and DQ_OBS_TABLES[key]:
        try:
            table_name = DQ_OBS_TABLES[key]
            if spark.catalog.tableExists(table_name):
                frame, origin, status, detail = spark.table(table_name), "DELTA:" + table_name, "DISPONIVEL", ""
            else:
                status, detail = "AUSENTE", "Tabela configurada não encontrada"
        except Exception as exc:
            status, detail = "ERRO", str(exc)[:300]
    if frame is not None:
        missing = spec["required"] - set(frame.columns)
        if missing:
            status, detail = "CONTRATO_INVALIDO", "Colunas ausentes: " + ", ".join(sorted(missing))
        else:
            if key in {"PLAN", "EXECUTION"} and DQ_OBS_RUN_ID is not None:
                frame = frame.where(F.col("run_id") == F.lit(str(DQ_OBS_RUN_ID)))
            try:
                rows = frame.limit(DQ_OBS_MAX_ROWS + 1).collect()
                if len(rows) > DQ_OBS_MAX_ROWS:
                    status, detail = "TRUNCADA", "Fonte excedeu o limite; nenhuma linha será utilizada"
                else:
                    dq_obs_data[key] = [r.asDict(recursive=True) for r in rows]
                    if not rows:
                        status, detail = "VAZIA", "Fonte disponível, mas sem registros"
            except Exception as exc:
                status, detail = "ERRO", str(exc)[:300]
    dq_obs_source_records.append((key, spec["variable"], origin, status, detail))

dq_obs_sources_df = dq_obs_frame(dq_obs_source_records, [
    ("fonte", T.StringType()), ("variavel", T.StringType()), ("origem", T.StringType()),
    ("status", T.StringType()), ("detalhe", T.StringType())
])
dq_obs_show(dq_obs_sources_df)
if "EXECUTION" not in dq_obs_data:
    print("ATENÇÃO: fonte EXECUTION indisponível; diagnósticos operacionais serão limitados.")
```

```python
# ============================================================
# 14.6 SELEÇÃO E NORMALIZAÇÃO DA EXECUÇÃO
# ============================================================

# O QUE FAZ:
# Normaliza o contrato operacional sem supor sucesso.

# COMO FAZ:
# Valida unicidade de run_id e converte campos com segurança.

# POR QUE É IMPORTANTE:
# Impede mistura de execuções.

# PERGUNTA RESPONDIDA:
# "Qual execução está sendo observada?"

raw_exec = dq_obs_data.get("EXECUTION", [])
raw_plan = dq_obs_data.get("PLAN", [])
run_ids = {str(r["run_id"]) for r in raw_exec + raw_plan if r.get("run_id") is not None}
if DQ_OBS_RUN_ID is None and len(run_ids) > 1:
    raise ValueError("Múltiplos run_id encontrados. Configure DQ_OBS_RUN_ID antes de continuar.")
DQ_OBS_ACTIVE_RUN_ID = str(DQ_OBS_RUN_ID) if DQ_OBS_RUN_ID is not None else next(iter(run_ids), None)
if DQ_OBS_ACTIVE_RUN_ID is not None:
    raw_exec = [r for r in raw_exec if str(r.get("run_id")) == DQ_OBS_ACTIVE_RUN_ID]
    raw_plan = [r for r in raw_plan if str(r.get("run_id")) == DQ_OBS_ACTIVE_RUN_ID]

execution_rows = []
for r in raw_exec:
    duration = r.get("duration_seconds")
    try:
        duration = float(duration) if duration is not None else None
    except (ValueError, TypeError):
        duration = None
    execution_rows.append((str(r.get("run_id")), str(r.get("profile")), str(r.get("status")),
                           str(r.get("started_at_utc")) if r.get("started_at_utc") is not None else None,
                           str(r.get("finished_at_utc")) if r.get("finished_at_utc") is not None else None,
                           duration, str(r.get("message"))[:1000] if r.get("message") is not None else None))
dq_obs_execution_df = dq_obs_frame(execution_rows, [
    ("run_id", T.StringType()), ("profile", T.StringType()), ("status", T.StringType()),
    ("inicio_utc", T.StringType()), ("fim_utc", T.StringType()),
    ("duracao_segundos", T.DoubleType()), ("mensagem", T.StringType())
])
dq_obs_show(dq_obs_execution_df)
```

```python
# ============================================================
# 14.7 SAÚDE DOS COMPONENTES
# ============================================================

# O QUE FAZ:
# Confronta plano e registros operacionais.

# COMO FAZ:
# Identifica sucesso, falha, bloqueio e ausência de registro.

# POR QUE É IMPORTANTE:
# Uma interrupção não pode ser interpretada como conclusão.

# PERGUNTA RESPONDIDA:
# "Quais componentes foram realmente concluídos?"

from collections import Counter
planned = {}
for r in raw_plan:
    key = str(r.get("profile"))
    if key in planned:
        raise ValueError("Profile duplicado no plano: " + key)
    planned[key] = r
observed = {}
for r in execution_rows:
    observed.setdefault(r[1], []).append(r)
health_rows = []
for profile in sorted(set(planned) | set(observed)):
    p = planned.get(profile)
    records = observed.get(profile, [])
    if len(records) > 1:
        state = "REGISTROS_DUPLICADOS"
    elif not records:
        state = "NAO_REGISTRADO"
    else:
        state = records[0][2]
    health_rows.append((DQ_OBS_ACTIVE_RUN_ID, profile, int(p["ordem"]) if p and p.get("ordem") is not None else None,
                        p is not None, bool(records), state, len(records)))
dq_obs_health_df = dq_obs_frame(health_rows, [
    ("run_id", T.StringType()), ("profile", T.StringType()), ("ordem", T.IntegerType()),
    ("planejado", T.BooleanType()), ("registrado", T.BooleanType()),
    ("estado", T.StringType()), ("quantidade_registros", T.IntegerType())
])
dq_obs_show(dq_obs_health_df)
```

```python
# ============================================================
# 14.8 MÉTRICAS DE DESEMPENHO
# ============================================================

# O QUE FAZ:
# Calcula duração observada por componente.

# COMO FAZ:
# Utiliza apenas tempos não negativos efetivamente registrados.

# POR QUE É IMPORTANTE:
# Destaca gargalos sem estimar valores ausentes.

# PERGUNTA RESPONDIDA:
# "Quais componentes excederam o limiar?"

performance_rows = []
for r in execution_rows:
    duration = r[5]
    valid = duration is not None and duration >= 0
    performance_rows.append((r[0], r[1], r[2], duration, valid,
                             bool(valid and duration > DQ_OBS_DURATION_WARNING_SECONDS)))
dq_obs_performance_df = dq_obs_frame(performance_rows, [
    ("run_id", T.StringType()), ("profile", T.StringType()), ("status", T.StringType()),
    ("duracao_segundos", T.DoubleType()), ("duracao_valida", T.BooleanType()),
    ("acima_limiar", T.BooleanType())
])
dq_obs_show(dq_obs_performance_df)
```

```python
# ============================================================
# 14.9 EVENTOS OPERACIONAIS
# ============================================================

# O QUE FAZ:
# Organiza falhas, bloqueios e inconsistências.

# COMO FAZ:
# Converte estados em eventos rastreáveis.

# POR QUE É IMPORTANTE:
# Facilita diagnóstico sem inferir causa raiz.

# PERGUNTA RESPONDIDA:
# "Quais eventos exigem investigação?"

event_rows = []
for run_id, profile, order, was_planned, registered, state, count in health_rows:
    if state == "SUCCESS" and was_planned:
        continue
    if state == "FAILED":
        severity, kind = "CRITICO", "FALHA_EXECUCAO"
    elif state in {"SKIPPED_DEPENDENCY", "NAO_REGISTRADO"}:
        severity, kind = "ALTO", "EXECUCAO_INCOMPLETA"
    elif state == "REGISTROS_DUPLICADOS":
        severity, kind = "ALTO", "REGISTRO_DUPLICADO"
    elif not was_planned:
        severity, kind = "MEDIO", "FORA_DO_PLANO"
    else:
        severity, kind = "MEDIO", "STATUS_DESCONHECIDO"
    messages = [r[6] for r in observed.get(profile, []) if r[6]]
    event_rows.append((run_id, profile, kind, severity, state, " | ".join(messages)[:1000] if messages else None))
for r in performance_rows:
    if not r[4]:
        event_rows.append((r[0], r[1], "DURACAO_INVALIDA", "MEDIO", r[2], "Duração ausente ou negativa"))
dq_obs_events_df = dq_obs_frame(event_rows, [
    ("run_id", T.StringType()), ("profile", T.StringType()), ("evento", T.StringType()),
    ("severidade", T.StringType()), ("estado", T.StringType()), ("mensagem", T.StringType())
])
dq_obs_show(dq_obs_events_df)
```

```python
# ============================================================
# 14.10 RASTREABILIDADE DA EXECUÇÃO
# ============================================================

# O QUE FAZ:
# Apresenta ordem planejada e estado registrado.

# COMO FAZ:
# Relaciona plano e execução por run_id e Profile.

# POR QUE É IMPORTANTE:
# Oferece rastreabilidade sem alegar tracing distribuído.

# PERGUNTA RESPONDIDA:
# "Qual foi a sequência planejada e seu resultado?"

trace_rows = []
for run_id, profile, order, was_planned, registered, state, count in health_rows:
    trace_rows.append((run_id, order, profile, state, was_planned, registered))
dq_obs_trace_df = dq_obs_frame(trace_rows, [
    ("run_id", T.StringType()), ("ordem_planejada", T.IntegerType()),
    ("profile", T.StringType()), ("estado", T.StringType()),
    ("planejado", T.BooleanType()), ("registrado", T.BooleanType())
])
dq_obs_show(dq_obs_trace_df.orderBy(F.col("ordem_planejada").asc_nulls_last(), "profile"))
```

```python
# ============================================================
# 14.11 DISPONIBILIDADE ANALÍTICA
# ============================================================

# O QUE FAZ:
# Consulta inventário opcional do módulo 12.

# COMO FAZ:
# Reutiliza apenas linhas de dq_report_sources_df disponíveis.

# POR QUE É IMPORTANTE:
# Separa disponibilidade de resultados da qualidade dos dados.

# PERGUNTA RESPONDIDA:
# "Quais dimensões possuem resultados consultáveis?"

analytical_rows = []
for r in dq_obs_data.get("REPORT_SOURCES", []):
    analytical_rows.append((str(r.get("dimensao")), str(r.get("status")),
                            str(r.get("origem")) if r.get("origem") is not None else None,
                            str(r.get("observacao"))[:500] if r.get("observacao") is not None else None))
dq_obs_analytical_sources_df = dq_obs_frame(analytical_rows, [
    ("dimensao", T.StringType()), ("status", T.StringType()),
    ("origem", T.StringType()), ("observacao", T.StringType())
])
dq_obs_show(dq_obs_analytical_sources_df)
```

```python
# ============================================================
# 14.12 CLASSIFICAÇÃO DE ALERTAS
# ============================================================

# O QUE FAZ:
# Consolida alertas explícitos sem envio externo.

# COMO FAZ:
# Aplica regras de status, duração e disponibilidade.

# POR QUE É IMPORTANTE:
# Permite priorizar investigação sem alterar Score.

# PERGUNTA RESPONDIDA:
# "Quais alertas devem ser investigados?"

alert_rows = []
for r in event_rows:
    alert_rows.append((r[0], r[1], r[3], r[2], r[5] or r[4]))
for r in performance_rows:
    if r[5]:
        alert_rows.append((r[0], r[1], "MEDIO", "DURACAO_ELEVADA",
                           "Duração acima do limite configurado"))
for r in dq_obs_source_records:
    if r[3] != "DISPONIVEL":
        level = "ALTO" if r[0] == "EXECUTION" else "MEDIO"
        alert_rows.append((DQ_OBS_ACTIVE_RUN_ID, r[0], level, "FONTE_" + r[3], r[4]))
for r in analytical_rows:
    if r[1] != "DISPONIVEL":
        alert_rows.append((DQ_OBS_ACTIVE_RUN_ID, r[0], "MEDIO", "RESULTADO_ANALITICO_" + r[1],
                           r[3] or "Fonte analítica não disponível integralmente"))
if not raw_plan:
    alert_rows.append((DQ_OBS_ACTIVE_RUN_ID, "PLAN", "MEDIO", "PLANO_INDISPONIVEL",
                       "Não é possível verificar se todos os Profiles planejados foram executados"))
dq_obs_alerts_df = dq_obs_frame(alert_rows, [
    ("run_id", T.StringType()), ("componente", T.StringType()),
    ("severidade", T.StringType()), ("tipo_alerta", T.StringType()),
    ("descricao", T.StringType())
])
dq_obs_show(dq_obs_alerts_df)
```

```python
# ============================================================
# 14.13 RESUMO EXECUTIVO
# ============================================================

# O QUE FAZ:
# Consolida métricas operacionais e ressalvas.

# COMO FAZ:
# Conta estados observados e duração válida.

# POR QUE É IMPORTANTE:
# Evita afirmar completude sem plano e fontes íntegros.

# PERGUNTA RESPONDIDA:
# "Qual é o panorama operacional da execução?"

states = Counter(r[5] for r in health_rows)
valid_durations = [r[3] for r in performance_rows if r[4]]
source_status = {r[0]: r[3] for r in dq_obs_source_records}
complete = bool(raw_plan) and source_status.get("PLAN") == "DISPONIVEL" and source_status.get("EXECUTION") == "DISPONIVEL" and all(
    r[3] and r[4] and r[5] == "SUCCESS" and r[6] == 1 for r in health_rows
)
summary_rows = [(
    DQ_OBS_ACTIVE_RUN_ID, len(planned), len(execution_rows), states["SUCCESS"],
    states["FAILED"], states["SKIPPED_DEPENDENCY"], states["NAO_REGISTRADO"],
    states["REGISTROS_DUPLICADOS"], float(sum(valid_durations)), len(valid_durations),
    len(alert_rows), complete, "COMPLETA_OPERACIONAL" if complete else "INDETERMINADA_OU_INCOMPLETA"
)]
dq_obs_summary_df = dq_obs_frame(summary_rows, [
    ("run_id", T.StringType()), ("profiles_planejados", T.IntegerType()),
    ("registros_execucao", T.IntegerType()), ("sucessos", T.IntegerType()),
    ("falhas", T.IntegerType()), ("bloqueados", T.IntegerType()),
    ("nao_registrados", T.IntegerType()), ("profiles_duplicados", T.IntegerType()),
    ("duracao_registrada_segundos", T.DoubleType()), ("duracoes_validas", T.IntegerType()),
    ("quantidade_alertas", T.IntegerType()), ("conclusao_operacional_completa", T.BooleanType()),
    ("interpretacao", T.StringType())
])
dq_obs_show(dq_obs_summary_df)
```

```python
# ============================================================
# 14.14 TESTES SINTÉTICOS E CONTRATOS
# ============================================================

# O QUE FAZ:
# Verifica contratos e casos de classificação sem reexecutar Profiles.

# COMO FAZ:
# Usa asserts sobre estruturas locais; nenhum job de fonte original.

# POR QUE É IMPORTANTE:
# Detecta regressões simples com baixo custo.

# PERGUNTA RESPONDIDA:
# "As saídas e regras básicas são consistentes?"

expected_outputs = {
    "dq_obs_sources_df": {"fonte", "status"},
    "dq_obs_execution_df": {"run_id", "profile", "status"},
    "dq_obs_health_df": {"profile", "estado", "planejado"},
    "dq_obs_performance_df": {"profile", "duracao_valida"},
    "dq_obs_events_df": {"evento", "severidade"},
    "dq_obs_trace_df": {"ordem_planejada", "profile"},
    "dq_obs_analytical_sources_df": {"dimensao", "status"},
    "dq_obs_alerts_df": {"tipo_alerta", "severidade"},
    "dq_obs_summary_df": {"run_id", "interpretacao"},
}
for variable, required in expected_outputs.items():
    assert isinstance(globals()[variable], DataFrame), variable
    assert required.issubset(set(globals()[variable].columns)), variable
assert len(health_rows) == len(set(planned) | set(observed))
assert all(r[3] >= 0 for r in performance_rows if r[4])
assert not complete or (len(raw_plan) > 0 and all(r[5] == "SUCCESS" for r in health_rows))
assert not (states["NAO_REGISTRADO"] and complete)
print("Contratos e verificações locais concluídos. Nenhum Profile foi executado.")
```
