#### CONSISTENCY PROFILE


##### 07.1 DOCUMENTAÇÃO DO CONSISTENCY PROFILE

###### Objetivo

O **Consistency Profile** tem como objetivo identificar, quantificar e analisar possíveis inconsistências presentes nos valores e nas relações entre as colunas do DataFrame.

Enquanto o **Schema Profile** analisa a estrutura das colunas, o **Distribution Profile** analisa a distribuição dos valores, o **Pattern Profile** identifica padrões estruturais e o **Duplicity Profile** avalia a repetição de informações, o **Consistency Profile** investiga a coerência interna dos dados.

A análise busca identificar situações em que valores ou relações apresentam comportamentos potencialmente incompatíveis com as características esperadas de consistência.

O framework não depende do significado de negócio das colunas.

O foco está nas características observáveis dos dados, como:

- consistência intrínseca dos valores;
- presença de textos vazios;
- espaços desnecessários nas extremidades;
- ocorrência de valores numéricos especiais;
- relações entre atributos;
- dependências funcionais candidatas;
- determinantes associados a múltiplos valores;
- consistência temporal;
- violações de regras explícitas;
- cobertura das análises;
- classificação das evidências identificadas.

###### Exemplos

**Exemplo 1 — Consistência intrínseca**

Considere os valores de uma coluna:

`ABC123`

` ABC123 `

`DEF456`

`""`

O segundo valor apresenta espaços nas extremidades, enquanto o último representa uma string vazia.

Essas características podem indicar problemas de padronização ou preenchimento.

Entretanto, a identificação dessas ocorrências não significa automaticamente que os valores sejam inválidos.

O **Consistency Profile** registra essas situações como sinais exploratórios.

**Exemplo 2 — Dependência funcional**

Considere os registros:

| Codigo | Descricao |
|---|---|
| 101 | Categoria A |
| 101 | Categoria A |
| 102 | Categoria B |
| 102 | Categoria C |
| 103 | Categoria D |

Nesse conjunto, o código `102` está associado a duas descrições diferentes.

Caso exista uma dependência funcional na qual:

`Codigo → Descricao`

cada valor de `Codigo` deveria determinar um único valor de `Descricao`.

Nesse exemplo, existe um determinante com múltiplos valores dependentes.

O framework identifica essa situação como uma possível inconsistência relacional.

Entretanto, a confirmação de uma violação depende da existência de uma regra que estabeleça essa relação.

**Exemplo 3 — Consistência temporal**

Considere os registros:

| Data_Inicio | Data_Fim |
|---|---|
| 2026-01-10 | 2026-01-12 |
| 2026-02-15 | 2026-02-14 |
| 2026-03-01 | 2026-03-05 |

Caso seja estabelecida a regra:

`Data_Inicio <= Data_Fim`

o segundo registro apresenta uma violação temporal.

Diferentemente de uma hipótese exploratória, essa ocorrência pode ser classificada como violação de regra explícita, pois existe uma condição previamente definida para a avaliação.

O **Consistency Profile** não presume automaticamente relações temporais com base nos nomes das colunas.

As regras devem ser configuradas de maneira explícita.

###### Principais análises

- Identificação de strings vazias;
- Identificação de espaços nas extremidades;
- Identificação de valores numéricos `NaN`;
- Identificação de valores infinitos;
- Análise de consistência intrínseca por coluna;
- Identificação de dependências funcionais candidatas;
- Contagem de determinantes distintos;
- Identificação de determinantes conflitantes;
- Quantificação de registros em grupos conflitantes;
- Avaliação de cobertura das relações analisadas;
- Verificação de regras temporais configuradas;
- Identificação de violações de regras explícitas;
- Classificação das evidências;
- Consolidação dos indicadores de consistência;
- Resumo executivo do Consistency Profile.

###### Interpretação dos resultados

O **Consistency Profile** diferencia evidências exploratórias de violações confirmadas por regras explícitas.

Essa separação é fundamental para evitar classificações incorretas sobre a qualidade dos dados.

As evidências são organizadas em duas categorias principais:

- **SINAL_EXPLORATORIO:** comportamento potencialmente inconsistente que necessita de interpretação adicional;
- **REGRA_EXPLICITA:** ocorrência que viola uma condição previamente configurada.

Uma dependência funcional candidata não representa necessariamente uma regra de negócio.

Da mesma forma, valores vazios, espaços externos e valores numéricos especiais podem apresentar significados diferentes conforme a origem e a finalidade dos dados.

Por esse motivo, o Profile não atribui automaticamente penalizações ou scores de qualidade às evidências identificadas.

A análise apresenta informações quantitativas que poderão subsidiar avaliações posteriores.

###### Característica

O **Consistency Profile** é independente da origem dos dados.

Sua análise é realizada exclusivamente sobre o DataFrame disponibilizado pelo **Data Preparation**.

O processamento utiliza operações distribuídas do Apache Spark para avaliar valores, agrupamentos, relações entre atributos e condições de consistência.

As análises relacionais utilizam limites configuráveis para controlar a quantidade de colunas e pares avaliados, evitando combinações excessivas e custos computacionais desnecessários.

As regras temporais são opcionais e devem ser definidas explicitamente.

As funções genéricas e reutilizáveis são mantidas no notebook de bibliotecas **00_FRAMEWORK_CONFIG**, enquanto o **Consistency Profile** concentra os parâmetros específicos, análises, consolidações e visualizações.

Dessa forma, o mesmo Profile pode ser utilizado em diferentes fontes, volumes e estruturas de dados.

```python
# ============================================================
# 07.3 PARÂMETROS DO PROFILE
# ============================================================

# O QUE FAZ:
# Define escopo, limites e regras explícitas.

# COMO FAZ:
# Configura colunas candidatas, número máximo de pares e regras temporais.

# POR QUE É IMPORTANTE:
# Mantém o Profile genérico e evita busca combinatória irrestrita.

# PERGUNTA RESPONDIDA:
# "Quais análises serão executadas?"

CONSISTENCY_IGNORE_BLANKS_IN_FD = True
CONSISTENCY_MAX_FD_COLUMNS = 8
CONSISTENCY_MAX_FD_PAIRS = 12
CONSISTENCY_FD_MIN_COVERAGE_PCT = 10.0
# Opcional: definir colunas específicas; None usa as primeiras colunas elegíveis.
CONSISTENCY_FD_COLUMNS = None
# Regras explícitas (início, fim). Vazio = nenhuma inferência temporal automática.
CONSISTENCY_TEMPORAL_PAIRS = []
CONSISTENCY_DISPLAY_RESULTS = True
```

```python
# ============================================================
# 07.4 VALIDAÇÃO E ELEGIBILIDADE
# ============================================================

# O QUE FAZ:
# Valida o DataFrame e a biblioteca.

# COMO FAZ:
# Verifica contrato, nomes duplicados e helpers de consistência.

# POR QUE É IMPORTANTE:
# Evita resultados incorretos por colunas ambíguas ou dependências ausentes.

# PERGUNTA RESPONDIDA:
# "O Profile pode executar com segurança?"

if "df_prepared" not in globals() or "total_registros" not in globals():
    raise ValueError("Execute o Data Preparation antes deste Profile.")
if len(df_prepared.columns) != len(set(df_prepared.columns)):
    raise ValueError("Nomes de colunas duplicados não são suportados.")
for helper in ("dq_col", "dq_valid_value", "dq_is_atomic", "dq_consistency_fd_aggregate"):
    if helper not in globals():
        raise ValueError(f"Função {helper} ausente em 00_FRAMEWORK_CONFIG; adicione o complemento da biblioteca.")
if CONSISTENCY_MAX_FD_PAIRS < 0 or CONSISTENCY_MAX_FD_COLUMNS < 0:
    raise ValueError("Limites de pares e colunas devem ser não negativos.")
schema_por_coluna = {f.name: f.dataType for f in df_prepared.schema.fields}
colunas_atomicas = [c for c, t in schema_por_coluna.items() if dq_is_atomic(t)]
print(f"Registros: {total_registros} | Colunas atômicas: {len(colunas_atomicas)}")
```

```python
# ============================================================
# 07.5 CONSISTÊNCIA INTRÍNSECA
# ============================================================

# O QUE FAZ:
# Mede sinais estruturais em textos e números.

# COMO FAZ:
# Executa agregações distribuídas em uma única projeção de métricas.

# POR QUE É IMPORTANTE:
# Separa anomalias observáveis de regras de negócio.

# PERGUNTA RESPONDIDA:
# "Quais colunas apresentam sinais estruturais de inconsistência?"

metricas = []
meta = []
for i, campo in enumerate(df_prepared.schema.fields):
    nome, tipo = campo.name, campo.dataType
    c = dq_col(nome)
    if isinstance(tipo, StringType):
        metricas.extend([
            F.sum(F.when(c.isNotNull() & (F.length(F.trim(c)) == 0), 1).otherwise(0)).alias(f"m_{i}_blank"),
            F.sum(F.when(c.isNotNull() & (c != F.trim(c)), 1).otherwise(0)).alias(f"m_{i}_spaces")
        ])
        meta.append((nome, "texto", f"m_{i}_blank", f"m_{i}_spaces", None, None))
    elif isinstance(tipo, (FloatType, DoubleType)):
        metricas.extend([
            F.sum(F.when(F.isnan(c), 1).otherwise(0)).alias(f"m_{i}_nan"),
            F.sum(F.when(c.isin(float('inf'), float('-inf')), 1).otherwise(0)).alias(f"m_{i}_inf")
        ])
        meta.append((nome, "numerico_ponto_flutuante", None, None, f"m_{i}_nan", f"m_{i}_inf"))
    else:
        meta.append((nome, "nao_aplicavel", None, None, None, None))
valores = df_prepared.agg(*metricas).first().asDict() if metricas else {}
linhas = []
for nome, categoria, blank, spaces, nan, inf in meta:
    b = int(valores.get(blank) or 0) if blank else None
    s = int(valores.get(spaces) or 0) if spaces else None
    na = int(valores.get(nan) or 0) if nan else None
    fi = int(valores.get(inf) or 0) if inf else None
    linhas.append((nome, categoria, b, s, na, fi))
schema_colunas = "coluna string, categoria string, textos_vazios long, espacos_externos long, valores_nan long, valores_infinitos long"
consistency_column_profile_df = spark.createDataFrame(linhas, schema_colunas)
if CONSISTENCY_DISPLAY_RESULTS:
    display(consistency_column_profile_df)
```

```python
# ============================================================
# 07.6 SELEÇÃO DE DEPENDÊNCIAS FUNCIONAIS
# ============================================================

# O QUE FAZ:
# Seleciona pares direcionais de atributos para exploração.

# COMO FAZ:
# Limita candidatos e exclui tipos complexos, sem inferir semântica por nomes.

# POR QUE É IMPORTANTE:
# Controla custo e torna a seleção reproduzível.

# PERGUNTA RESPONDIDA:
# "Quais hipóteses A -> B serão avaliadas?"

from itertools import permutations
if CONSISTENCY_FD_COLUMNS is None:
    candidatas = colunas_atomicas[:CONSISTENCY_MAX_FD_COLUMNS]
else:
    invalidas = [c for c in CONSISTENCY_FD_COLUMNS if c not in colunas_atomicas]
    if invalidas:
        raise ValueError(f"Colunas não elegíveis para FD: {invalidas}")
    candidatas = list(dict.fromkeys(CONSISTENCY_FD_COLUMNS))[:CONSISTENCY_MAX_FD_COLUMNS]
pares_fd = list(permutations(candidatas, 2))[:CONSISTENCY_MAX_FD_PAIRS]
print(f"Pares direcionais selecionados: {len(pares_fd)}")
print(pares_fd)
```

```python
# ============================================================
# 07.7 DEPENDÊNCIAS FUNCIONAIS CANDIDATAS
# ============================================================

# O QUE FAZ:
# Avalia conflitos A -> B nos registros elegíveis.

# COMO FAZ:
# Agrupa por determinante e conta distintos no atributo dependente.

# POR QUE É IMPORTANTE:
# Expõe relações suspeitas sem declarar erros de negócio.

# PERGUNTA RESPONDIDA:
# "Um valor de A se associa a múltiplos valores de B?"

resultados_fd = []
for determinante, dependente in pares_fd:
    tipo_a, tipo_b = schema_por_coluna[determinante], schema_por_coluna[dependente]
    elegiveis = df_prepared.where(
        dq_valid_value(determinante, tipo_a, CONSISTENCY_IGNORE_BLANKS_IN_FD)
        & dq_valid_value(dependente, tipo_b, CONSISTENCY_IGNORE_BLANKS_IN_FD)
    ).select(dq_col(determinante).alias("_a"), dq_col(dependente).alias("_b"))
    grupos = elegiveis.groupBy("_a").agg(
        F.count(F.lit(1)).alias("registros"),
        F.countDistinct("_b").alias("valores_dependentes_distintos")
    )
    r = dq_consistency_fd_aggregate(grupos).first()
    validos = int(r["registros_validos"] or 0)
    conflito = int(r["registros_em_grupos_conflitantes"] or 0)
    cobertura = 100.0 * validos / total_registros if total_registros else 0.0
    resultados_fd.append((determinante, dependente, validos,
        int(r["determinantes_distintos"] or 0), int(r["determinantes_conflitantes"] or 0),
        conflito, round(cobertura, 4), round(100.0 * conflito / validos, 4) if validos else None,
        "COBERTURA_INSUFICIENTE" if cobertura < CONSISTENCY_FD_MIN_COVERAGE_PCT else "HIPOTESE_EXPLORATORIA"))
schema_fd = "determinante string, dependente string, registros_validos long, determinantes_distintos long, determinantes_conflitantes long, registros_em_grupos_conflitantes long, cobertura_pct double, registros_conflitantes_pct double, classificacao string"
consistency_fd_profile_df = spark.createDataFrame(resultados_fd, schema_fd)
if CONSISTENCY_DISPLAY_RESULTS:
    display(consistency_fd_profile_df)
```

```python
# ============================================================
# 07.8 REGRAS TEMPORAIS EXPLÍCITAS
# ============================================================

# O QUE FAZ:
# Verifica ordem temporal apenas para pares declarados.

# COMO FAZ:
# Calcula comparações início > fim em linhas com ambos os valores preenchidos.

# POR QUE É IMPORTANTE:
# Evita inferir cronologia por nomes ou assumir regras inexistentes.

# PERGUNTA RESPONDIDA:
# "Quais registros violam as regras temporais configuradas?"

resultados_temporais = []
for inicio, fim in CONSISTENCY_TEMPORAL_PAIRS:
    if inicio not in schema_por_coluna or fim not in schema_por_coluna:
        raise ValueError(f"Par temporal contém coluna inexistente: {(inicio, fim)}")
    if not all(isinstance(schema_por_coluna[c], (DateType, TimestampType)) for c in (inicio, fim)):
        raise TypeError(f"Par temporal exige colunas date/timestamp: {(inicio, fim)}")
    a, b = dq_col(inicio), dq_col(fim)
    r = df_prepared.agg(
        F.sum(F.when(a.isNotNull() & b.isNotNull(), 1).otherwise(0)).alias("validos"),
        F.sum(F.when(a.isNotNull() & b.isNotNull() & (a > b), 1).otherwise(0)).alias("violacoes")
    ).first()
    validos, violacoes = int(r['validos'] or 0), int(r['violacoes'] or 0)
    resultados_temporais.append((inicio, fim, validos, violacoes,
        round(100.0 * violacoes / validos, 4) if validos else None, "REGRA_EXPLICITA"))
schema_temporal = "inicio string, fim string, registros_comparaveis long, violacoes long, violacoes_pct double, classificacao string"
consistency_temporal_profile_df = spark.createDataFrame(resultados_temporais, schema_temporal)
if CONSISTENCY_DISPLAY_RESULTS:
    display(consistency_temporal_profile_df)
```

```python
# ============================================================
# 07.9 CONSOLIDAÇÃO DO PROFILE
# ============================================================

# O QUE FAZ:
# Consolida evidências estruturais, relacionais e temporais.

# COMO FAZ:
# Padroniza dimensões e valores para integração futura.

# POR QUE É IMPORTANTE:
# Facilita observabilidade e Data Quality Score sem misturar semânticas.

# PERGUNTA RESPONDIDA:
# "Quais evidências de consistência foram encontradas?"

evidencias = []
for r in consistency_column_profile_df.collect():
    for nome_metrica in ("textos_vazios", "espacos_externos", "valores_nan", "valores_infinitos"):
        v = r[nome_metrica]
        if v is not None:
            evidencias.append(("INTRINSECA", r['coluna'], None, nome_metrica, int(v), "SINAL_EXPLORATORIO"))
for r in consistency_fd_profile_df.collect():
    evidencias.append(("RELACIONAL", r['determinante'], r['dependente'], "registros_em_grupos_conflitantes", int(r['registros_em_grupos_conflitantes']), r['classificacao']))
for r in consistency_temporal_profile_df.collect():
    evidencias.append(("TEMPORAL", r['inicio'], r['fim'], "violacoes", int(r['violacoes']), "REGRA_EXPLICITA"))
schema_evidencias = "dimensao string, coluna_origem string, coluna_relacionada string, metrica string, quantidade long, classificacao string"
consistency_profile_df = spark.createDataFrame(evidencias, schema_evidencias)
if CONSISTENCY_DISPLAY_RESULTS:
    display(consistency_profile_df)
```

```python
print("Quantidade de evidências:", len(evidencias))
print("Primeira evidência:", evidencias[0] if evidencias else "Nenhuma")
```

```python

# ============================================================
# 07.10 RESUMO EXECUTIVO
# ============================================================

# O QUE FAZ:
# Consolida os principais indicadores executivos do
# Consistency Profile.
#
# COMO FAZ:
# Percorre as evidências registradas e contabiliza
# separadamente os sinais exploratórios e as violações
# de regras explícitas.
#
# POR QUE É IMPORTANTE:
# Permite avaliar o panorama geral da consistência dos
# dados sem confundir sinais exploratórios com erros
# confirmados ou gerar um score arbitrário.
#
# PERGUNTA RESPONDIDA:
# "Qual é o panorama executivo da consistência?"

# ============================================================
# CONSOLIDAÇÃO DAS EVIDÊNCIAS
# ============================================================

# Estrutura de cada evidência:
# [0] categoria
# [1] coluna_origem
# [2] coluna_relacionada
# [3] metrica
# [4] quantidade
# [5] classificacao

total_sinais = sum(
    int(r[4])
    for r in evidencias
    if r[5] == "SINAL_EXPLORATORIO"
)

total_regras = sum(
    int(r[4])
    for r in evidencias
    if r[5] == "REGRA_EXPLICITA"
)

# ============================================================
# CONSTRUÇÃO DO RESUMO EXECUTIVO
# ============================================================

resumo = [(
    int(total_registros),
    int(len(df_prepared.columns)),
    int(len(colunas_atomicas)),
    int(len(pares_fd)),
    int(len(CONSISTENCY_TEMPORAL_PAIRS)),
    int(total_sinais),
    int(total_regras)
)]

consistency_summary_df = spark.createDataFrame(
    resumo,
    """
    total_registros LONG,
    total_colunas INT,
    colunas_atomicas INT,
    pares_fd_avaliados INT,
    regras_temporais_avaliadas INT,
    total_sinalizacoes_intrinsecas LONG,
    total_violacoes_regras_explicitas LONG
    """
)

# ============================================================
# APRESENTAÇÃO DOS RESULTADOS
# ============================================================

if CONSISTENCY_DISPLAY_RESULTS:
    display(consistency_summary_df)

print("=" * 60)
print("CONSISTENCY PROFILE - RESUMO EXECUTIVO")
print("=" * 60)

print(f"Total de registros: {total_registros:,}")
print(f"Total de colunas: {len(df_prepared.columns)}")
print(f"Colunas atômicas analisáveis: {len(colunas_atomicas)}")
print(f"Pares de dependência funcional avaliados: {len(pares_fd)}")
print(f"Regras temporais avaliadas: {len(CONSISTENCY_TEMPORAL_PAIRS)}")
print(f"Total de sinais exploratórios: {total_sinais:,}")
print(f"Total de violações de regras explícitas: {total_regras:,}")

print("-" * 60)
print(
    "Observação: sinais exploratórios não representam "
    "necessariamente erros confirmados de qualidade."
)
print(
    "As contagens podem incluir múltiplas sinalizações "
    "para um mesmo registro."
)
print("=" * 60)

# ============================================================
# VALIDAÇÃO DOS INDICADORES
# ============================================================

assert total_sinais >= 0, (
    "O total de sinais exploratórios não pode ser negativo."
)

assert total_regras >= 0, (
    "O total de violações de regras explícitas não pode ser negativo."
)

print("Consistency Profile concluído com sucesso.")
```

```python
# ============================================================
# 07.11 TESTES SINTÉTICOS
# ============================================================

# O QUE FAZ:
# Valida os indicadores essenciais com um dataset pequeno.

# COMO FAZ:
# Cria dados artificiais e compara contagens esperadas.

# POR QUE É IMPORTANTE:
# Reduz o risco de erros nas métricas e nos denominadores.

# PERGUNTA RESPONDIDA:
# "As métricas fundamentais respeitam exemplos controlados?"

teste = spark.createDataFrame([
    ("A", "X", " ok "), ("A", "Y", ""), ("B", "Z", "ok")
], "codigo string, descricao string, texto string")
assert teste.where(dq_col("texto") != F.trim(dq_col("texto"))).count() == 1
assert teste.where(F.length(F.trim(dq_col("texto"))) == 0).count() == 1
grupo_teste = teste.groupBy("codigo").agg(
    F.count(F.lit(1)).alias("registros"),
    F.countDistinct("descricao").alias("valores_dependentes_distintos"))
resultado_teste = dq_consistency_fd_aggregate(grupo_teste).first()
assert resultado_teste['determinantes_conflitantes'] == 1
assert resultado_teste['registros_em_grupos_conflitantes'] == 2
print("Testes sintéticos: OK")
```
