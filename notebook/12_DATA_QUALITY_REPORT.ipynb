#### DATA QUALITY REPORT


##### 12.1 DOCUMENTAÇÃO DO DATA QUALITY REPORT

###### Objetivo

O **Data Quality Report** é responsável por consolidar, organizar e apresentar os resultados produzidos pelos diferentes componentes analíticos do Data Quality Framework, utilizando recursos do **Databricks** e do **PySpark**.

Seu objetivo é disponibilizar uma visão integrada da qualidade observada nos dados, reunindo indicadores quantitativos, evidências exploratórias, cobertura das análises e resultados do Data Quality Score.

O Data Quality Report não realiza diretamente as análises estatísticas ou estruturais do dataset. Essas responsabilidades permanecem nos respectivos Profiles.

A análise busca responder:

> Quais dimensões de qualidade foram avaliadas, quais indicadores e evidências foram identificados, qual é a cobertura das análises e quais resultados devem orientar investigações posteriores?

O módulo possui caráter **analítico, descritivo e executivo**. Sua finalidade é facilitar a interpretação dos resultados, sem estabelecer automaticamente a aprovação ou reprovação do dataset.

---

###### Exemplos

**Exemplo 1 — Consolidação dos resultados de Schema**

Considere que o Schema Profile tenha identificado diferentes percentuais de preenchimento entre as colunas de um dataset.

O Data Quality Report recupera os indicadores disponíveis e os organiza em uma estrutura consolidada.

O resultado permite identificar atributos com menor completude observada, sem pressupor que todos os campos sejam obrigatórios.

**Exemplo 2 — Identificação de evidências exploratórias**

Considere que o Outlier Profile tenha identificado valores estatisticamente atípicos em determinadas colunas numéricas.

O Data Quality Report apresenta as sinalizações identificadas, preservando sua origem e natureza estatística.

Essas ocorrências não são classificadas automaticamente como erros, pois valores atípicos podem representar comportamentos legítimos dos dados.

**Exemplo 3 — Consolidação do Data Quality Score**

Considere que o módulo 10 tenha calculado um Score de 85 pontos, com cobertura de 60% das métricas catalogadas.

O Data Quality Report apresenta a pontuação juntamente com sua cobertura.

Nesse cenário, o Score representa o desempenho dos critérios efetivamente avaliados, enquanto a cobertura informa a abrangência dessa avaliação.

A pontuação isolada não deve ser interpretada como comprovação de qualidade global.

**Exemplo 4 — Ausência de resultados de um Profile**

Considere que o Correlation Profile não tenha sido executado ou que seus resultados não estejam disponíveis.

O Data Quality Report registra a indisponibilidade da fonte correspondente.

O módulo não atribui valores artificiais às métricas ausentes e não interpreta a ausência de resultados como ausência de problemas.

**Exemplo 5 — Consolidação de execuções independentes**

Considere que os Profiles tenham sido executados pelo Framework Orchestrator utilizando `dbutils.notebook.run()`.

Como os notebooks filhos possuem contextos isolados, os DataFrames criados durante essas execuções não são compartilhados automaticamente.

Nesse cenário, o Data Quality Report poderá utilizar resultados previamente persistidos em tabelas Delta, desde que suas fontes estejam configuradas.

---

###### Principais análises

O Data Quality Report contempla:

1. **Configuração do relatório**
   - Definição do modo de leitura;
   - Configuração das fontes disponíveis;
   - Definição dos limites de leitura;
   - Controle da exibição dos resultados;
   - Configuração da publicação opcional;
   - Identificação da execução quando disponível.

2. **Inventário das fontes**
   - Identificação dos Profiles esperados;
   - Verificação da disponibilidade dos resultados;
   - Identificação da origem dos DataFrames;
   - Registro de fontes indisponíveis;
   - Identificação de incompatibilidades;
   - Verificação de possíveis truncamentos.

3. **Validação dos contratos**
   - Verificação dos campos necessários;
   - Validação da estrutura dos resultados;
   - Identificação de campos ausentes;
   - Preservação da rastreabilidade;
   - Prevenção de interpretações inconsistentes.

4. **Consolidação dos indicadores**
   - Extração das métricas quantitativas disponíveis;
   - Identificação das dimensões analisadas;
   - Organização dos indicadores por origem;
   - Preservação das unidades e dos valores observados;
   - Diferenciação entre resultados disponíveis e ausentes.

5. **Consolidação das evidências**
   - Identificação de concentrações relevantes;
   - Apresentação de desvios de padrões;
   - Organização de evidências de duplicidade;
   - Consolidação de sinalizações de consistência;
   - Apresentação de outliers estatísticos;
   - Identificação de associações entre atributos.

6. **Integração do Data Quality Score**
   - Recuperação da pontuação calculada;
   - Apresentação da cobertura das métricas;
   - Apresentação da cobertura das dimensões;
   - Preservação dos critérios elegíveis;
   - Diferenciação entre Score e evidências exploratórias.

7. **Consolidação executiva**
   - Organização dos indicadores principais;
   - Identificação das dimensões disponíveis;
   - Apresentação das limitações da análise;
   - Construção de uma visão integrada;
   - Disponibilização de informações para acompanhamento.

8. **Recomendações e disponibilização**
   - Identificação de pontos que merecem investigação;
   - Registro de limitações metodológicas;
   - Organização dos resultados para consumo posterior;
   - Disponibilização de DataFrames padronizados;
   - Publicação opcional em tabelas Delta.

---

###### Métodos de execução

**Leitura em memória**

O modo `MEMORY` utiliza DataFrames disponíveis na sessão Python atual.

Essa estratégia é adequada quando os Profiles e o Data Quality Report compartilham o mesmo contexto de execução.

A existência de resultados em outros notebooks não garante que suas variáveis estejam disponíveis nesta sessão.

**Leitura em Delta**

O modo `DELTA` utiliza tabelas previamente persistidas e configuradas em `DQ_REPORT_TABLES`.

Essa estratégia permite consultar resultados de execuções anteriores sem reexecutar os Profiles.

As tabelas devem conter resultados agregados, e não os registros originais do dataset analisado.

**Leitura automática**

O modo `AUTO` prioriza DataFrames disponíveis em memória e utiliza tabelas Delta configuradas quando necessário.

A origem efetivamente utilizada deve ser registrada para permitir a rastreabilidade dos indicadores.

**Validação das fontes**

O módulo verifica a disponibilidade e a compatibilidade dos resultados antes de consolidá-los.

Fontes ausentes, incompatíveis ou truncadas devem ser identificadas, evitando a utilização silenciosa de informações incompletas.

**Consolidação dos resultados**

Os indicadores disponíveis são organizados em DataFrames específicos para inventário, métricas, evidências, resumo executivo e recomendações.

**Publicação opcional**

O módulo permite disponibilizar resultados consolidados em tabelas Delta quando a publicação estiver explicitamente habilitada.

A publicação não deve ocorrer automaticamente durante a execução padrão.

---

###### Tratamento de situações especiais

Para preservar a confiabilidade das informações, o Data Quality Report considera:

- **Fonte inexistente:** registrada como indisponível, sem geração de indicadores artificiais;
- **DataFrame incompatível:** identificado quando não apresenta os campos necessários;
- **Resultado vazio:** tratado sem pressupor ausência de problemas de qualidade;
- **Indicador nulo:** preservado como informação indisponível ou não calculada;
- **Resultado truncado:** identificado para impedir interpretações de cobertura completa;
- **Profile não executado:** não deve ser confundido com Profile executado sem sinalizações;
- **Execução isolada:** exige resultados acessíveis por memória compartilhada ou persistência;
- **Execuções diferentes:** não devem ser combinadas sem identificação e controle adequados;
- **Score indisponível:** não deve ser substituído por zero ou por uma pontuação presumida;
- **Publicação desabilitada:** os resultados permanecem disponíveis apenas no contexto de execução;
- **Ausência de todas as fontes:** deve produzir um diagnóstico de indisponibilidade, e não uma conclusão sobre a qualidade do dataset.

A disponibilidade de um indicador não garante, isoladamente, que ele seja suficiente para representar a qualidade global dos dados.

---

###### Interpretação dos resultados

Os resultados do Data Quality Report devem ser interpretados considerando a natureza de cada informação.

**Indicadores quantitativos**

Representam métricas calculadas pelos módulos anteriores, como percentuais de completude, contagens de sinalizações, cobertura e Score.

Esses indicadores devem preservar sua unidade de medida e seu contexto de cálculo.

**Evidências exploratórias**

Representam comportamentos observados que podem justificar investigações adicionais.

Uma concentração elevada, um padrão incomum, uma duplicidade candidata, um outlier ou uma correlação forte não representa necessariamente um erro.

**Data Quality Score**

Representa a consolidação dos critérios elegíveis definidos no módulo 10.

Sua interpretação deve considerar os pesos, a cobertura e a natureza dos critérios utilizados.

**Cobertura das análises**

Indica a abrangência das informações disponíveis e dos critérios avaliados.

Uma cobertura reduzida limita as conclusões que podem ser obtidas a partir do relatório.

**Fonte indisponível**

Indica que determinado resultado não pôde ser utilizado na consolidação.

Essa condição não representa aprovação nem reprovação da dimensão correspondente.

**Princípio fundamental:**

> A qualidade observada deve ser interpretada considerando os indicadores disponíveis, a cobertura das análises, a natureza das evidências e as limitações metodológicas.

O relatório não deve transformar automaticamente observações estatísticas em violações de qualidade.

---

###### Características do Data Quality Report

O módulo foi desenvolvido considerando os seguintes princípios:

- **Genérico:** não depende de regras específicas de negócio;
- **Reutilizável:** permite consolidar resultados de diferentes datasets;
- **Modular:** mantém a apresentação separada dos cálculos dos Profiles;
- **Configurável:** permite selecionar modos de leitura e fontes;
- **Rastreável:** preserva a origem dos indicadores;
- **Interpretável:** apresenta métricas e evidências com seu contexto;
- **Não destrutivo:** não modifica diretamente os dados analisados;
- **Conservador:** não substitui informações ausentes por valores artificiais;
- **Escalável:** prioriza resultados agregados;
- **Extensível:** permite incorporar novas fontes e indicadores;
- **Orientado à investigação:** apoia a identificação de pontos de atenção;
- **Independente do domínio:** não exige conhecimento prévio das regras de negócio;
- **Compatível com DataOps:** favorece acompanhamento, rastreabilidade e disponibilização de resultados.

O módulo deve permanecer responsável pela consolidação e apresentação, evitando assumir responsabilidades de ingestão, preparação ou processamento analítico.

---

###### DataFrames de resultado

O Data Quality Report disponibiliza estruturas padronizadas para consulta e integração.

**`dq_report_sources_df`**

Representa o inventário das fontes utilizadas no relatório.

Permite identificar disponibilidade, origem, compatibilidade e possíveis limitações dos resultados.

**`dq_report_kpis_df`**

Consolida os indicadores quantitativos recuperados dos Profiles.

Permite consultar métricas, valores observados, unidades e dimensões de origem.

**`dq_report_evidence_df`**

Organiza evidências exploratórias identificadas pelos módulos analíticos.

Permite investigar comportamentos observados sem classificá-los automaticamente como erros.

**`dq_report_executive_df`**

Disponibiliza uma visão executiva dos resultados consolidados.

Reúne informações relevantes sobre disponibilidade, cobertura e indicadores principais.

**`dq_report_recommendations_df`**

Registra recomendações condicionais, pontos de atenção e limitações metodológicas.

As recomendações devem ser interpretadas como orientações para investigação, não como decisões automáticas de correção.

**`dq_report_publication_df`**

Registra informações relacionadas à publicação opcional dos resultados.

Permite acompanhar a disponibilização das estruturas consolidadas quando essa funcionalidade estiver habilitada.

---

###### Limitações

A implementação atual apresenta algumas limitações que devem ser consideradas:

1. **Dependência dos Profiles:** o relatório depende da disponibilidade dos resultados previamente calculados.

2. **Isolamento de contexto:** DataFrames criados em notebooks executados isoladamente não são compartilhados automaticamente.

3. **Persistência não garantida:** a execução dos Profiles não assegura que seus resultados estejam disponíveis em tabelas Delta.

4. **Cobertura parcial:** a ausência de determinadas fontes pode limitar a abrangência do relatório.

5. **Possibilidade de truncamento:** limites configurados de leitura podem impedir a consolidação integral de determinadas fontes.

6. **Dependência dos contratos:** alterações nos schemas dos Profiles podem exigir adaptações no catálogo de fontes.

7. **Ausência de correção automática:** o módulo não modifica registros considerados problemáticos.

8. **Limitação do Score:** a pontuação representa somente os critérios efetivamente elegíveis e avaliados.

9. **Natureza exploratória:** determinadas evidências dependem de interpretação contextual para serem consideradas problemas de qualidade.

10. **Dependência do ambiente:** a leitura e a publicação dependem dos recursos disponíveis no Databricks.

11. **Comparabilidade entre execuções:** relatórios com diferentes critérios, parâmetros ou coberturas não devem ser comparados automaticamente.

12. **Ausência de aprovação universal:** o módulo não define um percentual único que determine a aprovação ou reprovação de qualquer dataset.

Essas limitações deverão orientar a evolução do framework nas etapas de persistência, observabilidade, governança e integração com soluções analíticas.

---

7. Qual é o Data Quality Score disponível?

Os resultados poderão ser utilizados posteriormente na construção de dashboards, relatórios gerenciais, históricos de qualidade e indicadores de acompanhamento associados às práticas de **DataOps** e **Data Governance**.

**Observação:** o Data Quality Report deve permanecer responsável pela consolidação e interpretação dos resultados. A definição de regras de negócio, critérios de aprovação e ações corretivas pertence às políticas de qualidade e aos processos de governança estabelecidos para os dados analisados.

```python
# ============================================================
# 12.3 CONFIGURAÇÕES DO RELATÓRIO
# ============================================================

# O QUE FAZ:
# Define modo, fontes, limites e publicação.

# COMO FAZ:
# Usa parâmetros explícitos e valida opções antes de ler dados.

# POR QUE É IMPORTANTE:
# Evita reprocessamento e escrita involuntária.

# PERGUNTA RESPONDIDA:
# "Quais resultados podem ser consultados?"

import json

DQ_REPORT_MODE = "MEMORY"  # MEMORY, DELTA ou AUTO
DQ_REPORT_RUN_ID = None    # Opcional; exige coluna run_id nas fontes
DQ_REPORT_MAX_ROWS_PER_SOURCE = 500
DQ_REPORT_MAX_EVIDENCES_PER_SOURCE = 25
DQ_REPORT_DISPLAY_RESULTS = True
DQ_REPORT_ENABLE_PUBLICATION = False
DQ_REPORT_OUTPUT_DATABASE = None  # Ex.: "catalogo.schema" (somente se publicação habilitada)
DQ_REPORT_OUTPUT_PREFIX = "dq_report"
DQ_REPORT_OUTPUT_WRITE_MODE = "errorifexists"  # Proteção contra sobrescrita

# Configure apenas tabelas Delta com perfis já materializados.
# None significa que nenhuma tabela foi configurada para a fonte.
DQ_REPORT_TABLES = {
    "SCHEMA": None,
    "DISTRIBUTION": None,
    "PATTERN": None,
    "DUPLICITY": None,
    "CONSISTENCY": None,
    "OUTLIER": None,
    "CORRELATION": None,
    "SCORE": None,
}

if DQ_REPORT_MODE not in {"MEMORY", "DELTA", "AUTO"}:
    raise ValueError("DQ_REPORT_MODE deve ser MEMORY, DELTA ou AUTO.")
if not isinstance(DQ_REPORT_MAX_ROWS_PER_SOURCE, int) or DQ_REPORT_MAX_ROWS_PER_SOURCE < 1:
    raise ValueError("DQ_REPORT_MAX_ROWS_PER_SOURCE deve ser inteiro >= 1.")
if not isinstance(DQ_REPORT_MAX_EVIDENCES_PER_SOURCE, int) or DQ_REPORT_MAX_EVIDENCES_PER_SOURCE < 1:
    raise ValueError("DQ_REPORT_MAX_EVIDENCES_PER_SOURCE deve ser inteiro >= 1.")
if DQ_REPORT_OUTPUT_WRITE_MODE not in {"errorifexists", "append"}:
    raise ValueError("Modo de publicação permitido: errorifexists ou append.")
```

```python
# ============================================================
# 12.4 CATÁLOGO E CONTRATOS DE ENTRADA
# ============================================================

# O QUE FAZ:
# Declara os perfis esperados e suas métricas conhecidas.

# COMO FAZ:
# Centraliza o mapeamento sem importar os notebooks anteriores.

# POR QUE É IMPORTANTE:
# Evita dependências ocultas e contratos divergentes.

# PERGUNTA RESPONDIDA:
# "Quais perfis e indicadores compõem o relatório?"

DQ_REPORT_CATALOG = {
    "SCHEMA": {
        "variable": "schema_profile_df", "required": ["coluna"],
        "metrics": {"pct_preenchimento": ("%", "DESCRITIVA"),
                    "pct_completude_efetiva": ("%", "DESCRITIVA")},
    },
    "DISTRIBUTION": {
        "variable": "distribution_summary_df", "required": ["coluna"],
        "metrics": {"concentracao_top1": ("%", "EXPLORATORIA"),
                    "concentracao_top3": ("%", "EXPLORATORIA"),
                    "concentracao_top5": ("%", "EXPLORATORIA")},
    },
    "PATTERN": {
        "variable": "pattern_summary_executivo_df", "required": ["coluna"],
        "metrics": {"pct_predominancia": ("%", "EXPLORATORIA"),
                    "pct_desvio_estrutura": ("%", "EXPLORATORIA"),
                    "pct_ausencia": ("%", "DESCRITIVA")},
    },
    "DUPLICITY": {
        "variable": "duplicity_summary_df", "required": [],
        "metrics": {"colunas_chave_simples": ("quantidade", "EXPLORATORIA"),
                    "pares_chave_composta": ("quantidade", "EXPLORATORIA"),
                    "pares_avaliados": ("quantidade", "DESCRITIVA")},
    },
    "CONSISTENCY": {
        "variable": "consistency_summary_df", "required": [],
        "metrics": {"total_sinalizacoes_intrinsecas": ("sinalizacoes", "EXPLORATORIA"),
                    "total_violacoes_regras_explicitas": ("violacoes", "REGRA_EXPLICITA"),
                    "regras_temporais_avaliadas": ("quantidade", "DESCRITIVA")},
    },
    "OUTLIER": {
        "variable": "outlier_summary_df", "required": [],
        "metrics": {"sinalizacoes_iqr": ("sinalizacoes_por_coluna", "EXPLORATORIA"),
                    "sinalizacoes_z": ("sinalizacoes_por_coluna", "EXPLORATORIA"),
                    "sinalizacoes_mad": ("sinalizacoes_por_coluna", "EXPLORATORIA")},
    },
    "CORRELATION": {
        "variable": "correlation_summary_df", "required": [],
        "metrics": {"avaliacoes_aplicaveis": ("quantidade", "DESCRITIVA"),
                    "avaliacoes_nao_aplicaveis": ("quantidade", "DESCRITIVA"),
                    "associacoes_fortes": ("pares_metodos", "EXPLORATORIA")},
    },
    "SCORE": {
        "variable": "dq_score_dataset_df", "required": ["score_geral"],
        "metrics": {"score_geral": ("pontos_0_100", "SCORE_OPERACIONAL"),
                    "cobertura_metricas_pct": ("%", "COBERTURA"),
                    "cobertura_dimensoes_pct": ("%", "COBERTURA")},
    },
}

assert set(DQ_REPORT_CATALOG) == set(DQ_REPORT_TABLES), "Catálogos de fontes divergentes."
```

```python
# ============================================================
# 12.5 LEITURA SEGURA DAS FONTES
# ============================================================

# O QUE FAZ:
# Obtém apenas perfis disponíveis e identifica fontes ausentes.

# COMO FAZ:
# Consulta variáveis em memória ou tabelas Delta explicitamente configuradas.

# POR QUE É IMPORTANTE:
# Impede reexecução automática dos módulos anteriores.

# PERGUNTA RESPONDIDA:
# "Quais resultados já existem e podem ser reutilizados?"

report_sources = {}
report_source_rows = []
report_execution_utc = datetime.now(timezone.utc).isoformat()

for dimension, config in DQ_REPORT_CATALOG.items():
    frame, origin, status, note = None, None, "AUSENTE", "Fonte não encontrada."
    variable = config["variable"]
    table_name = DQ_REPORT_TABLES[dimension]

    if DQ_REPORT_MODE in {"MEMORY", "AUTO"}:
        candidate = globals().get(variable)
        if isinstance(candidate, DataFrame):
            frame, origin, status, note = candidate, "MEMORY", "DISPONIVEL", ""

    if frame is None and DQ_REPORT_MODE in {"DELTA", "AUTO"} and table_name:
        try:
            if spark.catalog.tableExists(table_name):
                frame = spark.table(table_name)
                origin, status, note = f"DELTA:{table_name}", "DISPONIVEL", ""
            else:
                status, note = "AUSENTE", f"Tabela configurada não existe: {table_name}"
        except Exception as exc:
            status, note = "ERRO_LEITURA", str(exc)[:350]

    if frame is not None:
        missing = set(config["required"]) - set(frame.columns)
        if missing:
            status, note = "CONTRATO_INVALIDO", "Campos ausentes: " + ", ".join(sorted(missing))
        elif DQ_REPORT_RUN_ID is not None:
            if "run_id" not in frame.columns:
                status, note = "CONTRATO_INVALIDO", "Filtro run_id exigido, mas coluna ausente."
            else:
                frame = frame.where(F.col("run_id") == F.lit(DQ_REPORT_RUN_ID))

    if status == "DISPONIVEL":
        report_sources[dimension] = frame
    report_source_rows.append((dimension, variable, origin, status, note))

print("Fontes localizadas:", len(report_sources), "de", len(DQ_REPORT_CATALOG))
```

```python
# ============================================================
# 12.6 INVENTÁRIO DE COBERTURA
# ============================================================

# O QUE FAZ:
# Registra disponibilidade, quantidade amostrada e truncamento.

# COMO FAZ:
# Coleta somente um limite explícito de linhas dos perfis agregados.

# POR QUE É IMPORTANTE:
# Evita tratar amostras parciais como indicadores globais.

# PERGUNTA RESPONDIDA:
# "Qual é a cobertura real dos resultados disponíveis?"

report_samples = {}
source_rows = []
for dimension, variable, origin, status, note in report_source_rows:
    sample, truncated = [], False
    if status == "DISPONIVEL":
        try:
            sample = report_sources[dimension].limit(DQ_REPORT_MAX_ROWS_PER_SOURCE + 1).collect()
            truncated = len(sample) > DQ_REPORT_MAX_ROWS_PER_SOURCE
            sample = sample[:DQ_REPORT_MAX_ROWS_PER_SOURCE]
            if not sample:
                status, note = "SEM_REGISTROS", "Fonte existe, mas não retornou linhas."
            elif truncated:
                status, note = "TRUNCADA", "Limite de leitura atingido; indicadores não são completos."
        except Exception as exc:
            status, note = "ERRO_LEITURA", str(exc)[:350]
            sample = []
    report_samples[dimension] = sample
    source_rows.append((dimension, variable, origin, status, note, len(sample), bool(truncated)))

source_schema = """
    dimensao STRING, variavel STRING, origem STRING, status STRING,
    observacao STRING, linhas_lidas LONG, truncada BOOLEAN
"""
dq_report_sources_df = spark.createDataFrame(source_rows, schema=source_schema)
if DQ_REPORT_DISPLAY_RESULTS:
    display(dq_report_sources_df)
```

```python
# ============================================================
# 12.7 INDICADORES CONSOLIDADOS
# ============================================================

# O QUE FAZ:
# Publica valores conhecidos sem inventar métricas ausentes.

# COMO FAZ:
# Extrai apenas campos catalogados de fontes completas.

# POR QUE É IMPORTANTE:
# Diferencia ausência de resultado e valor observado igual a zero.

# PERGUNTA RESPONDIDA:
# "Quais indicadores foram efetivamente medidos?"

source_status = {row[0]: row[3] for row in source_rows}
kpi_rows = []
for dimension, config in DQ_REPORT_CATALOG.items():
    if source_status[dimension] != "DISPONIVEL":
        continue
    for metric, (unit, nature) in config["metrics"].items():
        if metric not in report_sources[dimension].columns:
            continue
        for row in report_samples[dimension]:
            value = row[metric]
            if value is None:
                continue
            if isinstance(value, bool):
                continue
            try:
                number = float(value)
                if not (number == number and abs(number) != float("inf")):
                    continue
            except (TypeError, ValueError, OverflowError):
                continue
            identifier = str(row["coluna"]) if "coluna" in row.asDict() and row["coluna"] is not None else None
            kpi_rows.append((dimension, identifier, metric, number, unit, nature))

kpi_schema = """
    dimensao STRING, coluna STRING, indicador STRING, valor DOUBLE,
    unidade STRING, natureza STRING
"""
dq_report_kpis_df = spark.createDataFrame(kpi_rows, schema=kpi_schema)
if DQ_REPORT_DISPLAY_RESULTS:
    display(dq_report_kpis_df)
```

```python
# ============================================================
# 12.8 EVIDÊNCIAS DE SCHEMA E DISTRIBUTION
# ============================================================

# O QUE FAZ:
# Organiza indicadores estruturais e de concentração.

# COMO FAZ:
# Seleciona os indicadores já disponíveis no catálogo de métricas.

# POR QUE É IMPORTANTE:
# Mantém interpretações descritivas e exploratórias separadas.

# PERGUNTA RESPONDIDA:
# "O que revelam completude e distribuição?"

# Não há novas agregações sobre dados originais.
report_schema_distribution_df = dq_report_kpis_df.where(
    F.col("dimensao").isin("SCHEMA", "DISTRIBUTION")
)
if DQ_REPORT_DISPLAY_RESULTS:
    display(report_schema_distribution_df)
```

```python
# ============================================================
# 12.9 EVIDÊNCIAS DE PATTERN E DUPLICITY
# ============================================================

# O QUE FAZ:
# Organiza indicadores de padrões e chaves candidatas.

# COMO FAZ:
# Filtra métricas já materializadas pelos Profiles.

# POR QUE É IMPORTANTE:
# Não classifica padrões raros nem duplicidade potencial como erro.

# PERGUNTA RESPONDIDA:
# "Quais características estruturais merecem atenção?"

report_pattern_duplicity_df = dq_report_kpis_df.where(
    F.col("dimensao").isin("PATTERN", "DUPLICITY")
)
if DQ_REPORT_DISPLAY_RESULTS:
    display(report_pattern_duplicity_df)
```

```python
# ============================================================
# 12.10 EVIDÊNCIAS DE CONSISTENCY E OUTLIER
# ============================================================

# O QUE FAZ:
# Reúne violações explícitas e sinalizações estatísticas.

# COMO FAZ:
# Filtra indicadores já consolidados e mantém a natureza de cada um.

# POR QUE É IMPORTANTE:
# Não confunde soma de sinalizações com registros únicos.

# PERGUNTA RESPONDIDA:
# "Quais evidências requerem verificação adicional?"

report_consistency_outlier_df = dq_report_kpis_df.where(
    F.col("dimensao").isin("CONSISTENCY", "OUTLIER")
)
if DQ_REPORT_DISPLAY_RESULTS:
    display(report_consistency_outlier_df)
```

```python
# ============================================================
# 12.11 EVIDÊNCIAS DE CORRELATION
# ============================================================

# O QUE FAZ:
# Disponibiliza associações e cobertura de métodos.

# COMO FAZ:
# Reutiliza o resumo de correlação já calculado.

# POR QUE É IMPORTANTE:
# Evita afirmar causalidade ou redundância funcional.

# PERGUNTA RESPONDIDA:
# "Quais associações foram sinalizadas?"

report_correlation_df = dq_report_kpis_df.where(
    F.col("dimensao") == "CORRELATION"
)
if DQ_REPORT_DISPLAY_RESULTS:
    display(report_correlation_df)
```

```python
# ============================================================
# 12.12 RESUMO EXECUTIVO
# ============================================================

# O QUE FAZ:
# Produz visão de disponibilidade, score e cobertura.

# COMO FAZ:
# Consolida informações já coletadas, sem varrer dados de origem.

# POR QUE É IMPORTANTE:
# Impede conclusões de aprovação quando o Score está ausente.

# PERGUNTA RESPONDIDA:
# "Qual é o panorama executivo da análise?"

from collections import Counter
status_counts = Counter(r[3] for r in source_rows)
score_values = {
    r[2]: r[3] for r in kpi_rows
    if r[0] == "SCORE" and r[1] is None
}
score_available = source_status.get("SCORE") == "DISPONIVEL"
executive_rows = [(
    report_execution_utc,
    DQ_REPORT_RUN_ID,
    len(DQ_REPORT_CATALOG),
    status_counts.get("DISPONIVEL", 0),
    len(DQ_REPORT_CATALOG) - status_counts.get("DISPONIVEL", 0),
    score_values.get("score_geral") if score_available else None,
    score_values.get("cobertura_metricas_pct") if score_available else None,
    score_values.get("cobertura_dimensoes_pct") if score_available else None,
    "PARCIAL" if status_counts.get("DISPONIVEL", 0) < len(DQ_REPORT_CATALOG) else "FONTES_DISPONIVEIS",
    "Score operacional não é certificação de qualidade ou aprovação do dataset."
)]
executive_schema = """
    gerado_em_utc STRING, run_id STRING, fontes_esperadas LONG,
    fontes_disponiveis LONG, fontes_indisponiveis_ou_parciais LONG,
    score_geral DOUBLE, cobertura_metricas_pct DOUBLE,
    cobertura_dimensoes_pct DOUBLE, status_cobertura STRING,
    ressalva STRING
"""
dq_report_executive_df = spark.createDataFrame(executive_rows, executive_schema)
if DQ_REPORT_DISPLAY_RESULTS:
    display(dq_report_executive_df)
```

```python
# ============================================================
# 12.13 RECOMENDAÇÕES E LIMITAÇÕES
# ============================================================

# O QUE FAZ:
# Registra próximos passos sem inventar diagnósticos.

# COMO FAZ:
# Gera recomendações condicionadas à disponibilidade e à natureza dos resultados.

# POR QUE É IMPORTANTE:
# Distingue ausência de evidência de ausência de problema.

# PERGUNTA RESPONDIDA:
# "O que deve ser investigado ou completado?"

recommendations = []
for dimension, variable, origin, status, note, read_rows, truncated in source_rows:
    if status != "DISPONIVEL":
        recommendations.append((
            dimension, "COBERTURA", "PENDENTE",
            "Disponibilizar perfil agregado compatível para esta dimensão.",
            f"Status: {status}. {note or ''}"[:500]
        ))

if source_status.get("SCORE") == "DISPONIVEL":
    recommendations.append((
        "SCORE", "INTERPRETACAO", "INFORMATIVO",
        "Avaliar score em conjunto com cobertura e critérios habilitados.",
        "Score não substitui regras explícitas de aceitação."
    ))

for dimension in ("DISTRIBUTION", "PATTERN", "DUPLICITY", "OUTLIER", "CORRELATION"):
    if source_status.get(dimension) == "DISPONIVEL":
        recommendations.append((
            dimension, "EVIDENCIA_EXPLORATORIA", "INFORMATIVO",
            "Investigar os sinais no contexto do domínio antes de classificá-los como erro.",
            "Indicadores exploratórios não devem reduzir o Score automaticamente."
        ))

recommendation_schema = """
    dimensao STRING, categoria STRING, prioridade STRING,
    recomendacao STRING, justificativa STRING
"""
dq_report_recommendations_df = spark.createDataFrame(recommendations, recommendation_schema)

# Uma linha por indicador conhecido e por coluna; evita a falsa promessa
# de reconstruir linhas de origem ou calcular anomalias inéditas.
evidence_rows = []
counts_by_dimension = Counter()
for dimension, column, metric, value, unit, nature in kpi_rows:
    if counts_by_dimension[dimension] >= DQ_REPORT_MAX_EVIDENCES_PER_SOURCE:
        continue
    evidence_rows.append((dimension, column, metric, value, unit, nature))
    counts_by_dimension[dimension] += 1

dq_report_evidence_df = spark.createDataFrame(evidence_rows, kpi_schema)
if DQ_REPORT_DISPLAY_RESULTS:
    display(dq_report_recommendations_df)
    display(dq_report_evidence_df)
```

```python
# ============================================================
# 12.14 PUBLICAÇÃO OPCIONAL
# ============================================================

# O QUE FAZ:
# Permite persistir apenas os resultados do relatório.

# COMO FAZ:
# Grava DataFrames consolidados em tabelas Delta somente mediante autorização explícita.

# POR QUE É IMPORTANTE:
# Evita sobrescritas acidentais e mantém publicação separada da análise.

# PERGUNTA RESPONDIDA:
# "Os artefatos do relatório devem ser publicados?"

report_outputs = {
    "sources": dq_report_sources_df,
    "kpis": dq_report_kpis_df,
    "evidence": dq_report_evidence_df,
    "executive": dq_report_executive_df,
    "recommendations": dq_report_recommendations_df,
}
publication_rows = []
if DQ_REPORT_ENABLE_PUBLICATION:
    if not DQ_REPORT_OUTPUT_DATABASE or not isinstance(DQ_REPORT_OUTPUT_DATABASE, str):
        raise ValueError("Configure DQ_REPORT_OUTPUT_DATABASE antes de publicar.")
    import re
    qualified_parts = DQ_REPORT_OUTPUT_DATABASE.split(".")
    if len(qualified_parts) not in (1, 2) or any(
        not re.fullmatch(r"[A-Za-z_][A-Za-z0-9_]*", part)
        for part in qualified_parts
    ):
        raise ValueError("Banco deve conter um ou dois identificadores simples separados por ponto.")
    if not re.fullmatch(r"[A-Za-z_][A-Za-z0-9_]*", DQ_REPORT_OUTPUT_PREFIX):
        raise ValueError("Prefixo de publicação inválido.")

    for name, frame in report_outputs.items():
        target = f"{DQ_REPORT_OUTPUT_DATABASE}.{DQ_REPORT_OUTPUT_PREFIX}_{name}"
        try:
            frame.write.format("delta").mode(DQ_REPORT_OUTPUT_WRITE_MODE).saveAsTable(target)
            publication_rows.append((name, target, "PUBLICADO", None))
        except Exception as exc:
            publication_rows.append((name, target, "FALHA", str(exc)[:500]))
else:
    for name in report_outputs:
        publication_rows.append((name, None, "DESABILITADO", None))

dq_report_publication_df = spark.createDataFrame(
    publication_rows,
    "artefato STRING, tabela STRING, status STRING, erro STRING"
)
if DQ_REPORT_DISPLAY_RESULTS:
    display(dq_report_publication_df)
print("Data Quality Report concluído. Nenhum Profile foi reexecutado.")
```
