##### 08.1 DOCUMENTAÇÃO DO OUTLIER PROFILE

###### Objetivo

O **Outlier Profile** é responsável por identificar, quantificar e caracterizar valores estatisticamente atípicos presentes nas colunas numéricas de um DataFrame, utilizando técnicas estatísticas implementadas com **PySpark**.

O objetivo é reconhecer comportamentos que se afastam significativamente da distribuição observada, sem depender do significado das colunas, da origem dos dados ou de regras específicas de negócio.

A análise busca responder:

> Existem valores que apresentam desvios estatisticamente relevantes em relação ao comportamento predominante dos dados?

O Profile tem caráter **exploratório e diagnóstico**. A identificação de um outlier não significa necessariamente que o valor esteja incorreto.

---

###### Exemplos

**Exemplo 1 — Identificação de valores extremos**

Considere uma coluna numérica com os seguintes valores:

```text
10, 11, 12, 12, 13, 14, 15, 150
```

O valor `150` apresenta um afastamento significativo em relação aos demais registros.

O Outlier Profile poderá sinalizar esse valor como estatisticamente atípico, permitindo investigar sua ocorrência.

**Exemplo 2 — Análise pelo intervalo interquartil (IQR)**

Considere uma distribuição com:

```text
Q1 = 20
Q3 = 40
IQR = 20

Limite inferior = 20 - (1,5 × 20) = -10
Limite superior = 40 + (1,5 × 20) = 70
```

Valores inferiores a `-10` ou superiores a `70` serão sinalizados pelo método IQR.

**Exemplo 3 — Comparação entre métodos**

Uma mesma coluna pode apresentar resultados diferentes conforme o método estatístico utilizado:

- **IQR:** avalia desvios em relação aos quartis.
- **Z-Score:** avalia a distância em relação à média, em unidades de desvio-padrão.
- **Modified Z-Score:** avalia a distância em relação à mediana, utilizando o desvio absoluto mediano (MAD).

A comparação permite identificar divergências e concordâncias entre os métodos, sem assumir que um deles representa a verdade absoluta.

---

###### Principais análises

O Outlier Profile contempla:

1. **Estatísticas descritivas**
   - Quantidade de valores válidos;
   - Média;
   - Mediana;
   - Desvio-padrão;
   - Valores mínimo e máximo.

2. **Análise de quartis**
   - Primeiro quartil (Q1);
   - Segundo quartil (Q2);
   - Terceiro quartil (Q3);
   - Intervalo interquartil (IQR).

3. **Detecção por IQR**
   - Cálculo dos limites inferior e superior;
   - Identificação de valores fora dos limites;
   - Quantidade e percentual de valores sinalizados.

4. **Detecção por Z-Score**
   - Cálculo do afastamento em relação à média;
   - Padronização pelo desvio-padrão;
   - Identificação de valores que ultrapassam o limiar configurado.

5. **Detecção por Modified Z-Score**
   - Cálculo da mediana;
   - Cálculo do MAD;
   - Identificação de valores extremos utilizando uma medida robusta de dispersão.

6. **Comparação dos métodos**
   - Quantidade de outliers por método;
   - Percentual de registros sinalizados;
   - Diferenças entre os detectores;
   - Identificação de métodos não aplicáveis.

7. **Consolidação executiva**
   - Quantidade de colunas analisadas;
   - Quantidade de métodos aplicados;
   - Total de sinalizações estatísticas;
   - Indicadores consolidados por coluna.

---

###### Tratamento de valores especiais

Para preservar a confiabilidade dos resultados, o Profile considera:

- **NULL:** excluído dos cálculos estatísticos;
- **NaN:** excluído das distribuições numéricas válidas;
- **Infinito positivo e negativo:** excluídos das estatísticas;
- **Desvio-padrão igual a zero:** Z-Score não aplicável;
- **MAD igual a zero:** Modified Z-Score não aplicável;
- **Colunas sem valores numéricos válidos:** classificadas como não analisáveis.

Os casos não aplicáveis devem ser diferenciados de situações em que um método foi executado e não identificou outliers.

---

###### Interpretação dos resultados

Os resultados representam **evidências estatísticas**, e não necessariamente problemas confirmados de qualidade.

Um valor extremo pode representar:

- Um erro de preenchimento ou processamento;
- Uma inconsistência na origem;
- Uma ocorrência rara, porém legítima;
- Uma característica natural da distribuição;
- Um evento excepcional que exige investigação.

A interpretação deve considerar a distribuição observada e, quando necessário, o contexto de negócio.

**Princípio fundamental:**

> Um outlier é uma observação estatisticamente atípica. Sua classificação como erro de qualidade depende de evidências adicionais.

Também é importante observar que a soma das sinalizações de diferentes métodos não representa necessariamente a quantidade de registros distintos com outliers, pois um mesmo registro pode ser sinalizado por mais de um detector.

---

###### Características do Profile

O Outlier Profile foi desenvolvido considerando os seguintes princípios:

- **Genérico:** não depende de regras específicas de negócio;
- **Reutilizável:** pode ser aplicado a diferentes DataFrames;
- **Modular:** possui responsabilidade analítica independente;
- **Configurável:** permite ajustar limiares e parâmetros estatísticos;
- **Escalável:** utiliza operações distribuídas do Apache Spark;
- **Robusto:** trata valores especiais e métodos não aplicáveis;
- **Explicável:** apresenta métricas que justificam as sinalizações;
- **Não destrutivo:** não altera nem remove registros do DataFrame original.

As operações analíticas são implementadas prioritariamente com **PySpark**, utilizando funções nativas de agregação, transformação e filtragem.

---

###### DataFrames de resultado

O Profile disponibiliza estruturas padronizadas para análise e integração com os demais componentes do framework.

**`outlier_profile_df`**

Consolida estatísticas descritivas, limites e indicadores de outliers por coluna.

**`outlier_method_comparison_df`**

Organiza os resultados por método estatístico, permitindo comparar quantidade de sinalizações, percentuais e aplicabilidade.

**`outlier_summary_df`**

Apresenta uma visão executiva das análises realizadas.

**`outlier_details_df`**

Disponibiliza exemplos limitados de registros sinalizados, quando a opção de detalhamento estiver habilitada.

---

###### Resultado esperado

Ao final da execução, o Outlier Profile deverá permitir responder:

1. Quais colunas numéricas apresentam valores estatisticamente atípicos?
2. Quantos registros foram sinalizados por cada método?
3. Qual percentual de valores foi sinalizado em cada coluna?
4. Quais são os limites estatísticos utilizados?
5. Quais métodos são aplicáveis a cada distribuição?
6. Existem diferenças relevantes entre IQR, Z-Score e Modified Z-Score?
7. Quais colunas merecem investigação prioritária?

Os resultados poderão ser utilizados posteriormente na consolidação das dimensões de qualidade, no **Data Quality Score** e em dashboards de monitoramento.

**Observação:** a identificação de outliers não implica exclusão, correção automática ou reprovação dos registros analisados.

````python
# ============================================================
# 08.3 PARÂMETROS DO PROFILE
# ============================================================
#
# O QUE FAZ:
# Define limites e custo computacional.
#
# COMO FAZ:
# Centraliza fatores, precisão, colunas e exibição.
#
# POR QUE É IMPORTANTE:
# Torna as análises reproduzíveis e ajustáveis.
#
# PERGUNTA RESPONDIDA:
# "Quais critérios estatísticos serão aplicados?"

OUTLIER_IQR_FACTOR = 1.5
OUTLIER_Z_THRESHOLD = 3.0
OUTLIER_MODIFIED_Z_THRESHOLD = 3.5
OUTLIER_PERCENTILE_ACCURACY = 10000
OUTLIER_MAX_COLUMNS = 20             # None para todas as numéricas
OUTLIER_EXCLUDE_COLUMNS = []          # Ex.: IDs numéricos, se conhecidos
OUTLIER_DISPLAY_RESULTS = True
OUTLIER_SHOW_DETAILS = False
OUTLIER_DETAILS_LIMIT = 20            # Por coluna e por método

assert OUTLIER_IQR_FACTOR > 0
assert OUTLIER_Z_THRESHOLD > 0
assert OUTLIER_MODIFIED_Z_THRESHOLD > 0
assert OUTLIER_PERCENTILE_ACCURACY >= 100
assert OUTLIER_MAX_COLUMNS is None or OUTLIER_MAX_COLUMNS > 0
assert OUTLIER_DETAILS_LIMIT > 0
````

````python
# ============================================================
# 08.4 VALIDAÇÃO E COLUNAS ELEGÍVEIS
# ============================================================
#
# O QUE FAZ:
# Verifica contrato e seleciona colunas numéricas.
#
# COMO FAZ:
# Utiliza df_prepared e colunas_outlier; filtra tipos reais.
#
# POR QUE É IMPORTANTE:
# Evita aplicar estatísticas em campos não numéricos.
#
# PERGUNTA RESPONDIDA:
# "Quais colunas podem ser analisadas?"

if "df_prepared" not in globals() or "colunas_outlier" not in globals():
    raise ValueError("Execute 02_DATA_PREPARATION antes do Profile 08.")

funcoes_necessarias = (
    "dq_outlier_col", "dq_outlier_valid",
    "dq_outlier_iqr_bounds", "dq_outlier_modified_z_threshold"
)
ausentes = [nome for nome in funcoes_necessarias if nome not in globals()]
if ausentes:
    raise ValueError("Adicione as funções do módulo 08 ao 00_FRAMEWORK_CONFIG: " + ", ".join(ausentes))

tipos = {campo.name: campo.dataType for campo in df_prepared.schema.fields}
colunas_elegiveis = [
    nome for nome in colunas_outlier
    if nome in tipos and isinstance(tipos[nome], NumericType)
    and nome not in OUTLIER_EXCLUDE_COLUMNS
]
colunas_elegiveis = list(dict.fromkeys(colunas_elegiveis))
if OUTLIER_MAX_COLUMNS is not None:
    colunas_elegiveis = colunas_elegiveis[:OUTLIER_MAX_COLUMNS]

print(f"Registros: {total_registros:,} | Colunas numéricas selecionadas: {len(colunas_elegiveis)}")
print("Colunas:", colunas_elegiveis)
````

````python
# ============================================================
# 08.5 ESTATÍSTICAS DESCRITIVAS E QUARTIS
# ============================================================
#
# O QUE FAZ:
# Calcula medidas de posição e dispersão por coluna.
#
# COMO FAZ:
# Usa agregações Spark e percentile_approx, sem dados brutos no driver.
#
# POR QUE É IMPORTANTE:
# Fornece referência estatística para os detectores.
#
# PERGUNTA RESPONDIDA:
# "Qual é o comportamento numérico de cada atributo?"

estatisticas = {}
for nome in colunas_elegiveis:
    coluna = dq_outlier_col(nome)
    valor = coluna.cast("double")
    base_valida = df_prepared.where(dq_outlier_valid(coluna)).select(valor.alias("valor"))

    agregados = base_valida.agg(
        F.count("valor").alias("n"),
        F.avg("valor").alias("media"),
        F.stddev_pop("valor").alias("desvio_padrao"),
        F.min("valor").alias("minimo"),
        F.max("valor").alias("maximo"),
        F.percentile_approx(
            "valor", [0.25, 0.50, 0.75], OUTLIER_PERCENTILE_ACCURACY
        ).alias("quartis")
    ).first()  # Apenas uma linha de métricas; os registros ficam no Spark.

    quartis = agregados["quartis"] or [None, None, None]
    q1, mediana, q3 = quartis
    inferior, superior, iqr = dq_outlier_iqr_bounds(q1, q3, OUTLIER_IQR_FACTOR)
    estatisticas[nome] = {
        "n": int(agregados["n"]),
        "media": agregados["media"],
        "desvio": agregados["desvio_padrao"],
        "minimo": agregados["minimo"],
        "maximo": agregados["maximo"],
        "q1": q1, "mediana": mediana, "q3": q3,
        "iqr": iqr, "limite_inferior": inferior,
        "limite_superior": superior
    }
print(f"Estatísticas calculadas para {len(estatisticas)} colunas.")
````

````python
# ============================================================
# 08.6 MAD ROBUSTO
# ============================================================
#
# O QUE FAZ:
# Calcula a mediana dos desvios absolutos por coluna.
#
# COMO FAZ:
# Agrega percentile_approx(|valor - mediana|, 0.5) em Spark.
#
# POR QUE É IMPORTANTE:
# Permite detectar extremos sem depender da média.
#
# PERGUNTA RESPONDIDA:
# "Qual é a dispersão robusta dos valores?"

for nome, s in estatisticas.items():
    if s["n"] == 0 or s["mediana"] is None:
        s["mad"] = None
        continue

    coluna = dq_outlier_col(nome)
    desvios_df = (
        df_prepared.where(dq_outlier_valid(coluna))
        .select(F.abs(coluna.cast("double") - F.lit(float(s["mediana"]))).alias("desvio_absoluto"))
    )
    s["mad"] = desvios_df.agg(
        F.percentile_approx(
            "desvio_absoluto", 0.5, OUTLIER_PERCENTILE_ACCURACY
        ).alias("mad")
    ).first()["mad"]  # Apenas um valor agregado por coluna.

print("MAD calculado via agregações distribuídas.")
````

````python
# ============================================================
# 08.7 DETECÇÃO DISTRIBUÍDA
# ============================================================
#
# O QUE FAZ:
# Conta sinalizações e concordância dos três métodos.
#
# COMO FAZ:
# Constrói flags Spark e agrega com F.sum(F.when(...)).
#
# POR QUE É IMPORTANTE:
# Evita loops Python sobre registros e distingue métodos inaplicáveis.
#
# PERGUNTA RESPONDIDA:
# "Quantos valores são sinalizados por cada método?"

resultados = []
for nome, s in estatisticas.items():
    n = s["n"]
    iqr_ok = n > 0 and s["iqr"] is not None and s["iqr"] > 0
    z_ok = n > 0 and s["desvio"] is not None and s["desvio"] > 0
    mad_limite = dq_outlier_modified_z_threshold(s["mad"], OUTLIER_MODIFIED_Z_THRESHOLD)
    mad_ok = n > 0 and mad_limite is not None
    aplicabilidade = {"iqr": iqr_ok, "z": z_ok, "mad": mad_ok}

    if n == 0:
        resultados.append((nome, s, aplicabilidade, None))
        continue

    coluna = dq_outlier_col(nome)
    x = coluna.cast("double")
    fi = ((x < F.lit(s["limite_inferior"])) | (x > F.lit(s["limite_superior"]))) if iqr_ok else F.lit(False)
    fz = (F.abs(x - F.lit(s["media"])) > F.lit(OUTLIER_Z_THRESHOLD * s["desvio"])) if z_ok else F.lit(False)
    fm = (F.abs(x - F.lit(s["mediana"])) > F.lit(mad_limite)) if mad_ok else F.lit(False)

    contagem = (
        df_prepared.where(dq_outlier_valid(coluna))
        .agg(
            F.sum(F.when(fi, 1).otherwise(0)).alias("iqr"),
            F.sum(F.when(fz, 1).otherwise(0)).alias("z"),
            F.sum(F.when(fm, 1).otherwise(0)).alias("mad"),
            F.sum(F.when(fi | fz | fm, 1).otherwise(0)).alias("uniao"),
            F.sum(F.when(fi & fz & fm, 1).otherwise(0)).alias("intersecao_tres")
        ).first()
    )
    contagens = {k: int(contagem[k] or 0) for k in ("iqr", "z", "mad", "uniao", "intersecao_tres")}
    resultados.append((nome, s, aplicabilidade, contagens))

print("Contagens obtidas por agregações PySpark.")
````

````python
# ============================================================
# 08.8 PERFIL CONSOLIDADO
# ============================================================
#
# O QUE FAZ:
# Produz indicadores estatísticos e de aplicabilidade.
#
# COMO FAZ:
# Cria DataFrame Spark tipado com métricas agregadas.
#
# POR QUE É IMPORTANTE:
# Permite consumo padronizado por Score e dashboards.
#
# PERGUNTA RESPONDIDA:
# "Qual é o perfil estatístico de cada coluna?"

schema_profile = """coluna string, total_registros long, registros_validos long, registros_excluidos long,
media double, desvio_padrao double, minimo double, maximo double, q1 double, mediana double,
q3 double, iqr double, mad double, limite_inferior_iqr double, limite_superior_iqr double,
iqr_aplicavel boolean, z_aplicavel boolean, mad_aplicavel boolean,
outliers_iqr long, outliers_z long, outliers_mad long, outliers_uniao long,
outliers_intersecao_tres long, pct_iqr double, pct_z double, pct_mad double, pct_uniao double"""

linhas = []
for nome, s, a, c in resultados:
    n = s["n"]
    c = c or {}
    def percentual(chave, aplicavel=True):
        return round(100.0 * c[chave] / n, 4) if n > 0 and aplicavel and chave in c else None

    linhas.append((
        nome, int(total_registros), n, int(total_registros) - n,
        s["media"], s["desvio"], s["minimo"], s["maximo"],
        s["q1"], s["mediana"], s["q3"], s["iqr"], s["mad"],
        s["limite_inferior"], s["limite_superior"],
        a["iqr"], a["z"], a["mad"],
        c.get("iqr") if a["iqr"] else None,
        c.get("z") if a["z"] else None,
        c.get("mad") if a["mad"] else None,
        c.get("uniao") if c else None,
        c.get("intersecao_tres") if all(a.values()) else None,
        percentual("iqr", a["iqr"]), percentual("z", a["z"]),
        percentual("mad", a["mad"]), percentual("uniao")
    ))

outlier_profile_df = spark.createDataFrame(linhas, schema_profile)
if OUTLIER_DISPLAY_RESULTS:
    display(outlier_profile_df)
````

````python
# ============================================================
# 08.9 COMPARAÇÃO DOS MÉTODOS
# ============================================================
#
# O QUE FAZ:
# Normaliza as métricas por coluna e detector.
#
# COMO FAZ:
# Usa stack sobre o DataFrame Spark de perfil.
#
# POR QUE É IMPORTANTE:
# Facilita comparação e visualização de desempenho dos detectores.
#
# PERGUNTA RESPONDIDA:
# "Quais métodos sinalizam mais valores?"

outlier_method_comparison_df = (
    outlier_profile_df.selectExpr(
        "coluna", "registros_validos",
        "stack(3, "
        "'IQR', iqr_aplicavel, outliers_iqr, pct_iqr, "
        "'Z_SCORE', z_aplicavel, outliers_z, pct_z, "
        "'MODIFIED_Z_SCORE', mad_aplicavel, outliers_mad, pct_mad"
        ") as (metodo, aplicavel, outliers, percentual)"
    )
)
if OUTLIER_DISPLAY_RESULTS:
    display(outlier_method_comparison_df)
````

````python
# ============================================================
# 08.10 RESUMO EXECUTIVO
# ============================================================
#
# O QUE FAZ:
# Consolida cobertura e sinalizações.
#
# COMO FAZ:
# Agrega o perfil por meio de funções Spark.
#
# POR QUE É IMPORTANTE:
# Evita confundir soma por coluna com registros únicos.
#
# PERGUNTA RESPONDIDA:
# "Qual é o panorama geral da detecção?"

outlier_summary_df = outlier_profile_df.agg(
    F.count("coluna").cast("long").alias("colunas_avaliadas"),
    F.sum(F.when(F.col("registros_validos") > 0, 1).otherwise(0)).cast("long").alias("colunas_com_valores_validos"),
    F.sum(F.coalesce(F.col("outliers_iqr"), F.lit(0))).cast("long").alias("sinalizacoes_iqr"),
    F.sum(F.coalesce(F.col("outliers_z"), F.lit(0))).cast("long").alias("sinalizacoes_z"),
    F.sum(F.coalesce(F.col("outliers_mad"), F.lit(0))).cast("long").alias("sinalizacoes_mad"),
    F.sum(F.coalesce(F.col("outliers_uniao"), F.lit(0))).cast("long").alias("sinalizacoes_uniao_por_coluna")
).withColumn("total_registros", F.lit(int(total_registros)).cast("long")) .withColumn("total_colunas", F.lit(len(df_prepared.columns)).cast("int"))
# Sem colunas elegíveis, cria uma linha de resumo com zeros.
if not colunas_elegiveis:
    outlier_summary_df = spark.createDataFrame(
        [(int(total_registros), len(df_prepared.columns), 0, 0, 0, 0, 0, 0)],
        "total_registros long, total_colunas int, colunas_avaliadas long, "
        "colunas_com_valores_validos long, sinalizacoes_iqr long, sinalizacoes_z long, "
        "sinalizacoes_mad long, sinalizacoes_uniao_por_coluna long"
    )
if OUTLIER_DISPLAY_RESULTS:
    display(outlier_summary_df)
print("Observação: as somas são sinalizações por coluna, não registros únicos globais.")
````

````python
# ============================================================
# 08.11 DETALHES OPCIONAIS
# ============================================================
#
# O QUE FAZ:
# Apresenta exemplos limitados de valores sinalizados.
#
# COMO FAZ:
# Filtra e limita DataFrames Spark por coluna e método.
#
# POR QUE É IMPORTANTE:
# Permite investigar exemplos sem coletar todo o dataset.
#
# PERGUNTA RESPONDIDA:
# "Quais valores foram sinalizados?"

outlier_details_df = spark.createDataFrame([], "coluna string, valor string, metodo string")
if OUTLIER_SHOW_DETAILS:
    amostras = []
    for nome, s, a, _ in resultados:
        coluna = dq_outlier_col(nome)
        x = coluna.cast("double")
        base = df_prepared.where(dq_outlier_valid(coluna))
        filtros = []
        if a["iqr"]:
            filtros.append(("IQR", (x < F.lit(s["limite_inferior"])) | (x > F.lit(s["limite_superior"]))))
        if a["z"]:
            filtros.append(("Z_SCORE", F.abs(x - F.lit(s["media"])) > F.lit(OUTLIER_Z_THRESHOLD * s["desvio"])))
        if a["mad"]:
            limite = dq_outlier_modified_z_threshold(s["mad"], OUTLIER_MODIFIED_Z_THRESHOLD)
            filtros.append(("MODIFIED_Z_SCORE", F.abs(x - F.lit(s["mediana"])) > F.lit(limite)))
        for metodo, condicao in filtros:
            amostras.append(
                base.where(condicao)
                .select(F.lit(nome).alias("coluna"), coluna.cast("string").alias("valor"), F.lit(metodo).alias("metodo"))
                .limit(OUTLIER_DETAILS_LIMIT)
            )
    for amostra in amostras:
        outlier_details_df = outlier_details_df.unionByName(amostra)
    if OUTLIER_DISPLAY_RESULTS:
        display(outlier_details_df)
````

````python
# ============================================================
# 08.12 TESTES SINTÉTICOS PYSPARK
# ============================================================
#
# O QUE FAZ:
# Verifica casos constantes, extremos e valores inválidos.
#
# COMO FAZ:
# Usa DataFrames Spark pequenos e asserts sobre agregados.
#
# POR QUE É IMPORTANTE:
# Reduz risco de erros matemáticos e regressões.
#
# PERGUNTA RESPONDIDA:
# "Os critérios funcionam em cenários controlados?"

teste_df = spark.createDataFrame(
    [(1.0,), (1.0,), (1.0,), (1.0,), (100.0,), (None,), (float("nan"),), (float("inf"),)],
    "valor double"
)
teste_valido = teste_df.where(dq_outlier_valid(F.col("valor")))
teste_stats = teste_valido.agg(
    F.count("valor").alias("n"),
    F.min("valor").alias("minimo"),
    F.max("valor").alias("maximo")
).first()
assert teste_stats["n"] == 5
assert teste_stats["minimo"] == 1.0
assert teste_stats["maximo"] == 100.0
assert dq_outlier_iqr_bounds(2.0, 6.0, 1.5) == (-4.0, 12.0, 4.0)
assert dq_outlier_modified_z_threshold(0.0) is None
assert dq_outlier_modified_z_threshold(2.0) > 0
print("Testes básicos concluídos com sucesso.")
````
