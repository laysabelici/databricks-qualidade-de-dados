##### 08.1 DOCUMENTAÇÃO DO OUTLIER PROFILE

###### Objetivo

O **Outlier Profile** é responsável por identificar, quantificar e caracterizar valores estatisticamente atípicos presentes nas colunas numéricas de um DataFrame, utilizando técnicas estatísticas implementadas com **PySpark**.

O objetivo é reconhecer comportamentos que se afastam significativamente da distribuição observada, sem depender do significado das colunas, da origem dos dados ou de regras específicas de negócio.

A análise busca responder:

> Existem valores que apresentam desvios estatisticamente relevantes em relação ao comportamento predominante dos dados?

O Profile tem caráter **exploratório e diagnóstico**. A identificação de um outlier não significa necessariamente que o valor esteja incorreto.

---

###### Exemplos

**Exemplo 1 — Identificação de valores extremos**

Considere uma coluna numérica com os seguintes valores:

```text
10, 11, 12, 12, 13, 14, 15, 150
```

O valor `150` apresenta um afastamento significativo em relação aos demais registros.

O Outlier Profile poderá sinalizar esse valor como estatisticamente atípico, permitindo investigar sua ocorrência.

**Exemplo 2 — Análise pelo intervalo interquartil (IQR)**

Considere uma distribuição com:

```text
Q1 = 20
Q3 = 40
IQR = 20

Limite inferior = 20 - (1,5 × 20) = -10
Limite superior = 40 + (1,5 × 20) = 70
```

Valores inferiores a `-10` ou superiores a `70` serão sinalizados pelo método IQR.

**Exemplo 3 — Comparação entre métodos**

Uma mesma coluna pode apresentar resultados diferentes conforme o método estatístico utilizado:

- **IQR:** avalia desvios em relação aos quartis.
- **Z-Score:** avalia a distância em relação à média, em unidades de desvio-padrão.
- **Modified Z-Score:** avalia a distância em relação à mediana, utilizando o desvio absoluto mediano (MAD).

A comparação permite identificar divergências e concordâncias entre os métodos, sem assumir que um deles representa a verdade absoluta.

---

###### Principais análises

O Outlier Profile contempla:

1. **Estatísticas descritivas**
   - Quantidade de valores válidos;
   - Média;
   - Mediana;
   - Desvio-padrão;
   - Valores mínimo e máximo.

2. **Análise de quartis**
   - Primeiro quartil (Q1);
   - Segundo quartil (Q2);
   - Terceiro quartil (Q3);
   - Intervalo interquartil (IQR).

3. **Detecção por IQR**
   - Cálculo dos limites inferior e superior;
   - Identificação de valores fora dos limites;
   - Quantidade e percentual de valores sinalizados.

4. **Detecção por Z-Score**
   - Cálculo do afastamento em relação à média;
   - Padronização pelo desvio-padrão;
   - Identificação de valores que ultrapassam o limiar configurado.

5. **Detecção por Modified Z-Score**
   - Cálculo da mediana;
   - Cálculo do MAD;
   - Identificação de valores extremos utilizando uma medida robusta de dispersão.

6. **Comparação dos métodos**
   - Quantidade de outliers por método;
   - Percentual de registros sinalizados;
   - Diferenças entre os detectores;
   - Identificação de métodos não aplicáveis.

7. **Consolidação executiva**
   - Quantidade de colunas analisadas;
   - Quantidade de métodos aplicados;
   - Total de sinalizações estatísticas;
   - Indicadores consolidados por coluna.

---

###### Tratamento de valores especiais

Para preservar a confiabilidade dos resultados, o Profile considera:

- **NULL:** excluído dos cálculos estatísticos;
- **NaN:** excluído das distribuições numéricas válidas;
- **Infinito positivo e negativo:** excluídos das estatísticas;
- **Desvio-padrão igual a zero:** Z-Score não aplicável;
- **MAD igual a zero:** Modified Z-Score não aplicável;
- **Colunas sem valores numéricos válidos:** classificadas como não analisáveis.

Os casos não aplicáveis devem ser diferenciados de situações em que um método foi executado e não identificou outliers.

---

###### Interpretação dos resultados

Os resultados representam **evidências estatísticas**, e não necessariamente problemas confirmados de qualidade.

Um valor extremo pode representar:

- Um erro de preenchimento ou processamento;
- Uma inconsistência na origem;
- Uma ocorrência rara, porém legítima;
- Uma característica natural da distribuição;
- Um evento excepcional que exige investigação.

A interpretação deve considerar a distribuição observada e, quando necessário, o contexto de negócio.

**Princípio fundamental:**

> Um outlier é uma observação estatisticamente atípica. Sua classificação como erro de qualidade depende de evidências adicionais.

Também é importante observar que a soma das sinalizações de diferentes métodos não representa necessariamente a quantidade de registros distintos com outliers, pois um mesmo registro pode ser sinalizado por mais de um detector.

---

###### Características do Profile

O Outlier Profile foi desenvolvido considerando os seguintes princípios:

- **Genérico:** não depende de regras específicas de negócio;
- **Reutilizável:** pode ser aplicado a diferentes DataFrames;
- **Modular:** possui responsabilidade analítica independente;
- **Configurável:** permite ajustar limiares e parâmetros estatísticos;
- **Escalável:** utiliza operações distribuídas do Apache Spark;
- **Robusto:** trata valores especiais e métodos não aplicáveis;
- **Explicável:** apresenta métricas que justificam as sinalizações;
- **Não destrutivo:** não altera nem remove registros do DataFrame original.

As operações analíticas são implementadas prioritariamente com **PySpark**, utilizando funções nativas de agregação, transformação e filtragem.

---

###### DataFrames de resultado

O Profile disponibiliza estruturas padronizadas para análise e integração com os demais componentes do framework.

**`outlier_profile_df`**

Consolida estatísticas descritivas, limites e indicadores de outliers por coluna.

**`outlier_method_comparison_df`**

Organiza os resultados por método estatístico, permitindo comparar quantidade de sinalizações, percentuais e aplicabilidade.

**`outlier_summary_df`**

Apresenta uma visão executiva das análises realizadas.

**`outlier_details_df`**

Disponibiliza exemplos limitados de registros sinalizados, quando a opção de detalhamento estiver habilitada.

---

###### Resultado esperado

Ao final da execução, o Outlier Profile deverá permitir responder:

1. Quais colunas numéricas apresentam valores estatisticamente atípicos?
2. Quantos registros foram sinalizados por cada método?
3. Qual percentual de valores foi sinalizado em cada coluna?
4. Quais são os limites estatísticos utilizados?
5. Quais métodos são aplicáveis a cada distribuição?
6. Existem diferenças relevantes entre IQR, Z-Score e Modified Z-Score?
7. Quais colunas merecem investigação prioritária?

Os resultados poderão ser utilizados posteriormente na consolidação das dimensões de qualidade, no **Data Quality Score** e em dashboards de monitoramento.

**Observação:** a identificação de outliers não implica exclusão, correção automática ou reprovação dos registros analisados.

````python
# ============================================================
# 08.3 PARÂMETROS DO PROFILE
# ============================================================

# O QUE FAZ:
# Define os parâmetros estatísticos e operacionais do Outlier Profile.

# COMO FAZ:
# Configura fatores, limiares, precisão aproximada dos percentis, limites de colunas, exibição e testes opcionais.

# POR QUE É IMPORTANTE:
# Permite controlar custo, reprodutibilidade e interpretação dos detectores.

# PERGUNTA RESPONDIDA:
# "Quais critérios estatísticos e limites serão utilizados?"

OUTLIER_IQR_FACTOR = 1.5
OUTLIER_Z_THRESHOLD = 3.0
OUTLIER_MODIFIED_Z_THRESHOLD = 3.5
OUTLIER_PERCENTILE_ACCURACY = 10000
OUTLIER_MAX_COLUMNS = 30  # None para todas as colunas numéricas
OUTLIER_EXCLUDE_COLUMNS = []  # Ex.: identificadores numéricos conhecidos
OUTLIER_DISPLAY_RESULTS = True
OUTLIER_SHOW_DETAILS = False
OUTLIER_DETAILS_LIMIT = 100

assert OUTLIER_IQR_FACTOR > 0
assert OUTLIER_Z_THRESHOLD > 0
assert OUTLIER_MODIFIED_Z_THRESHOLD > 0
assert OUTLIER_PERCENTILE_ACCURACY >= 100
````

````python
# ============================================================
# 08.4 VALIDAÇÃO E COLUNAS ELEGÍVEIS
# ============================================================

# O QUE FAZ:
# Valida o contrato de entrada e seleciona atributos numéricos.

# COMO FAZ:
# Verifica DataFrame, colunas, parâmetros e funções compartilhadas, excluindo tipos não numéricos.

# POR QUE É IMPORTANTE:
# Evita executar métodos estatísticos sobre atributos incompatíveis ou configurações inválidas.

# PERGUNTA RESPONDIDA:
# "Quais colunas numéricas estão elegíveis para análise?"

if "df_prepared" not in locals() or "colunas_outlier" not in locals():
    raise ValueError("Execute 02_DATA_PREPARATION antes deste Profile.")
if "total_registros" not in locals():
    raise ValueError("total_registros não foi disponibilizado.")
for fn in ("dq_outlier_col", "dq_outlier_valid", "dq_outlier_iqr_bounds", "dq_outlier_modified_z_threshold"):
    if fn not in globals():
        raise ValueError(f"Função {fn} ausente: adicione o complemento 08 à biblioteca central.")

if colunas_outlier is None:
    colunas_outlier = [campo.name for campo in df_prepared.schema.fields
                       if isinstance(campo.dataType, NumericType)]
colunas_elegiveis = [c for c in colunas_outlier if c not in OUTLIER_EXCLUDE_COLUMNS]
if OUTLIER_MAX_COLUMNS is not None:
    colunas_elegiveis = colunas_elegiveis[:OUTLIER_MAX_COLUMNS]
print(f"Registros: {total_registros:,} | Colunas numéricas selecionadas: {len(colunas_elegiveis)}")
````

````python
# ============================================================
# 08.5 ESTATÍSTICAS DESCRITIVAS E QUARTIS
# ============================================================

# O QUE FAZ:
# Calcula medidas de posição e dispersão das colunas elegíveis.

# COMO FAZ:
# Utiliza agregações distribuídas e percentis aproximados sobre valores numéricos finitos.

# POR QUE É IMPORTANTE:
# Fornece parâmetros de referência aos métodos IQR, Z-Score e Modified Z-Score.

# PERGUNTA RESPONDIDA:
# "Como se distribuem os valores válidos de cada coluna?"

estatisticas = {}
for nome in colunas_elegiveis:
    x = dq_outlier_col(nome).cast("double")
    validos = df_prepared.where(dq_outlier_valid(dq_outlier_col(nome))).select(x.alias("valor"))
    linha = validos.agg(
        F.count("valor").alias("n"),
        F.avg("valor").alias("media"),
        F.stddev_samp("valor").alias("desvio_padrao"),
        F.min("valor").alias("minimo"),
        F.max("valor").alias("maximo"),
        F.expr(f"percentile_approx(valor, array(0.25D, 0.5D, 0.75D), {OUTLIER_PERCENTILE_ACCURACY})").alias("quartis")
    ).first()
    q = linha["quartis"] or [None, None, None]
    inferior, superior, iqr = dq_outlier_iqr_bounds(q[0], q[2], OUTLIER_IQR_FACTOR)
    estatisticas[nome] = {
        "n": int(linha["n"]), "media": linha["media"], "desvio": linha["desvio_padrao"],
        "minimo": linha["minimo"], "maximo": linha["maximo"],
        "q1": q[0], "mediana": q[1], "q3": q[2],
        "iqr": iqr, "limite_inferior": inferior, "limite_superior": superior
    }
````

````python
# ============================================================
# 08.6 DESVIO ABSOLUTO MEDIANO (MAD)
# ============================================================

# O QUE FAZ:
# Calcula a dispersão robusta em relação à mediana.

# COMO FAZ:
# Aplica percentile_approx aos desvios absolutos dos valores válidos.

# POR QUE É IMPORTANTE:
# Permite avaliar afastamentos robustos sem depender exclusivamente da média.

# PERGUNTA RESPONDIDA:
# "Qual é a dispersão robusta de cada coluna numérica?"

for nome, s in estatisticas.items():
    if s["n"] == 0 or s["mediana"] is None:
        s["mad"] = None
        continue
    x = dq_outlier_col(nome).cast("double")
    s["mad"] = df_prepared.where(dq_outlier_valid(dq_outlier_col(nome))).select(
        F.abs(x - F.lit(float(s["mediana"]))).alias("desvio_absoluto")
    ).agg(F.expr(
        f"percentile_approx(desvio_absoluto, 0.5D, {OUTLIER_PERCENTILE_ACCURACY})"
    ).alias("mad")).first()["mad"]
````

````python

# ============================================================
# 08.7 DETECÇÃO DISTRIBUÍDA DE OUTLIERS
# ============================================================

# O QUE FAZ:
# Identifica e quantifica valores estatisticamente atípicos
# utilizando IQR, Z-Score e Modified Z-Score.

# COMO FAZ:
# Verifica a aplicabilidade dos métodos, constrói condições
# distribuídas com PySpark e calcula as contagens por coluna.

# POR QUE É IMPORTANTE:
# Permite comparar os detectores sem confundir métodos
# não aplicáveis com métodos que não identificaram outliers.
# Também mantém um contrato consistente com a célula 08.8.

# PERGUNTA RESPONDIDA:
# "Quantos valores são sinalizados por cada método estatístico?"

resultados = []

for nome, s in estatisticas.items():

    n = int(s["n"])

    # --------------------------------------------------------
    # 1. VERIFICAÇÃO DA APLICABILIDADE
    # --------------------------------------------------------

    iqr_ok = (
        n > 0
        and s["iqr"] is not None
        and s["iqr"] > 0
    )

    z_ok = (
        n > 0
        and s["desvio"] is not None
        and s["desvio"] > 0
    )

    mad_limite = dq_outlier_modified_z_threshold(
        s.get("mad"),
        OUTLIER_MODIFIED_Z_THRESHOLD
    )

    mad_ok = (
        n > 0
        and mad_limite is not None
    )

    aplicabilidade = {
        "iqr": bool(iqr_ok),
        "z": bool(z_ok),
        "mad": bool(mad_ok)
    }

    # --------------------------------------------------------
    # 2. TRATAMENTO DE MÉTODOS NÃO APLICÁVEIS
    # --------------------------------------------------------

    if n == 0 or not any(aplicabilidade.values()):

        resultados.append((
            nome,
            s,
            aplicabilidade,
            None
        ))

        continue

    # --------------------------------------------------------
    # 3. CONSTRUÇÃO DAS CONDIÇÕES ESTATÍSTICAS
    # --------------------------------------------------------

    coluna = dq_outlier_col(nome)
    x = coluna.cast("double")

    flag_iqr = (
        (
            (x < F.lit(s["limite_inferior"]))
            | (x > F.lit(s["limite_superior"]))
        )
        if iqr_ok
        else F.lit(False)
    )

    flag_z = (
        (
            F.abs(x - F.lit(s["media"]))
            > F.lit(OUTLIER_Z_THRESHOLD * s["desvio"])
        )
        if z_ok
        else F.lit(False)
    )

    flag_mad = (
        (
            F.abs(x - F.lit(s["mediana"]))
            > F.lit(mad_limite)
        )
        if mad_ok
        else F.lit(False)
    )

    # --------------------------------------------------------
    # 4. CONTAGEM DISTRIBUÍDA DAS SINALIZAÇÕES
    # --------------------------------------------------------

    base_valida = df_prepared.where(
        dq_outlier_valid(coluna)
    )

    agregados = base_valida.agg(
        F.sum(
            F.when(flag_iqr, 1).otherwise(0)
        ).alias("iqr"),

        F.sum(
            F.when(flag_z, 1).otherwise(0)
        ).alias("z"),

        F.sum(
            F.when(flag_mad, 1).otherwise(0)
        ).alias("mad"),

        F.sum(
            F.when(
                flag_iqr | flag_z | flag_mad,
                1
            ).otherwise(0)
        ).alias("uniao"),

        F.sum(
            F.when(
                flag_iqr & flag_z & flag_mad,
                1
            ).otherwise(0)
        ).alias("intersecao_tres")
    ).first()

    # --------------------------------------------------------
    # 5. PADRONIZAÇÃO DAS CONTAGENS
    # --------------------------------------------------------

    contagens = {
        chave: int(agregados[chave] or 0)
        for chave in (
            "iqr",
            "z",
            "mad",
            "uniao",
            "intersecao_tres"
        )
    }

    # --------------------------------------------------------
    # 6. CONSOLIDAÇÃO DO RESULTADO
    # --------------------------------------------------------

    resultados.append((
        nome,
        s,
        aplicabilidade,
        contagens
    ))

# ------------------------------------------------------------
# 7. VALIDAÇÃO DO CONTRATO DE SAÍDA
# ------------------------------------------------------------

assert all(
    len(resultado) == 4
    for resultado in resultados
), "Erro: os resultados devem possuir quatro elementos."

print("=" * 60)
print("OUTLIER PROFILE - DETECÇÃO DISTRIBUÍDA")
print("=" * 60)

print(f"Colunas processadas: {len(resultados)}")

print(
    "Colunas com pelo menos um método aplicável:",
    sum(
        1
        for _, _, aplicabilidade, _ in resultados
        if any(aplicabilidade.values())
    )
)

print(
    "Colunas sem métodos aplicáveis:",
    sum(
        1
        for _, _, aplicabilidade, _ in resultados
        if not any(aplicabilidade.values())
    )
)

print("Contrato de saída validado com sucesso.")
````

````python
# ============================================================
# 08.8 PERFIL CONSOLIDADO
# ============================================================

# O QUE FAZ:
# Consolida estatísticas descritivas, limites estatísticos,
# aplicabilidade e contagens de outliers por coluna.

# COMO FAZ:
# Utiliza os quatro elementos produzidos pela célula 08.7:
# nome, estatísticas, aplicabilidade e contagens.
# Constrói um DataFrame Spark com schema explícito.

# POR QUE É IMPORTANTE:
# Garante consistência entre as etapas do framework,
# diferencia métodos não aplicáveis de métodos sem outliers
# e preserva os indicadores para análises posteriores.

# PERGUNTA RESPONDIDA:
# "Qual é o perfil estatístico de outliers de cada coluna?"

schema_profile = """
    coluna STRING,
    total_registros LONG,
    registros_validos LONG,
    registros_excluidos LONG,
    media DOUBLE,
    desvio_padrao DOUBLE,
    minimo DOUBLE,
    maximo DOUBLE,
    q1 DOUBLE,
    mediana DOUBLE,
    q3 DOUBLE,
    iqr DOUBLE,
    mad DOUBLE,
    limite_inferior_iqr DOUBLE,
    limite_superior_iqr DOUBLE,
    iqr_aplicavel BOOLEAN,
    z_aplicavel BOOLEAN,
    mad_aplicavel BOOLEAN,
    outliers_iqr LONG,
    outliers_z LONG,
    outliers_mad LONG,
    outliers_uniao LONG,
    outliers_intersecao_tres LONG,
    pct_iqr DOUBLE,
    pct_z DOUBLE,
    pct_mad DOUBLE,
    pct_uniao DOUBLE
"""

# ------------------------------------------------------------
# 1. CONSTRUÇÃO DOS REGISTROS CONSOLIDADOS
# ------------------------------------------------------------

linhas = []

for nome, s, aplicabilidade, contagens in resultados:

    n = int(s["n"])
    c = contagens if contagens is not None else {}

    iqr_ok = bool(aplicabilidade["iqr"])
    z_ok = bool(aplicabilidade["z"])
    mad_ok = bool(aplicabilidade["mad"])

    algum_metodo_aplicavel = (
        iqr_ok or z_ok or mad_ok
    )

    todos_metodos_aplicaveis = (
        iqr_ok and z_ok and mad_ok
    )

    # --------------------------------------------------------
    # 2. CÁLCULO DOS PERCENTUAIS
    # --------------------------------------------------------

    def percentual(metodo, aplicavel):

        if not aplicavel or n == 0:
            return None

        quantidade = c.get(metodo)

        if quantidade is None:
            return None

        return round(
            100.0 * int(quantidade) / n,
            4
        )

    # --------------------------------------------------------
    # 3. CONSOLIDAÇÃO DA LINHA
    # --------------------------------------------------------

    linhas.append((
        str(nome),
        int(total_registros),
        n,
        int(total_registros) - n,

        s["media"],
        s["desvio"],
        s["minimo"],
        s["maximo"],
        s["q1"],
        s["mediana"],
        s["q3"],
        s["iqr"],
        s.get("mad"),
        s["limite_inferior"],
        s["limite_superior"],

        iqr_ok,
        z_ok,
        mad_ok,

        c.get("iqr") if iqr_ok else None,
        c.get("z") if z_ok else None,
        c.get("mad") if mad_ok else None,

        (
            c.get("uniao")
            if algum_metodo_aplicavel
            else None
        ),

        (
            c.get("intersecao_tres")
            if todos_metodos_aplicaveis
            else None
        ),

        percentual("iqr", iqr_ok),
        percentual("z", z_ok),
        percentual("mad", mad_ok),
        percentual("uniao", algum_metodo_aplicavel)
    ))

# ------------------------------------------------------------
# 4. CRIAÇÃO DO DATAFRAME SPARK
# ------------------------------------------------------------

outlier_profile_df = spark.createDataFrame(
    linhas,
    schema=schema_profile
)

# ------------------------------------------------------------
# 5. EXIBIÇÃO DOS RESULTADOS
# ------------------------------------------------------------

if OUTLIER_DISPLAY_RESULTS:
    display(outlier_profile_df)

# ------------------------------------------------------------
# 6. VALIDAÇÃO FINAL
# ------------------------------------------------------------

print("=" * 60)
print("OUTLIER PROFILE - PERFIL CONSOLIDADO")
print("=" * 60)

print(f"Colunas consolidadas: {len(linhas)}")
print(f"Indicadores disponíveis: {len(outlier_profile_df.columns)}")
print("Perfil consolidado construído com sucesso.")
````

````python
# ============================================================
# 08.9 COMPARAÇÃO DOS MÉTODOS
# ============================================================

# O QUE FAZ:
# Organiza os indicadores por coluna e método estatístico.

# COMO FAZ:
# Reestrutura o perfil consolidado em linhas para IQR, Z-Score e Modified Z-Score.

# POR QUE É IMPORTANTE:
# Facilita a comparação de cobertura, aplicabilidade e quantidade de sinalizações.

# PERGUNTA RESPONDIDA:
# "Como os resultados diferem entre os três métodos?"

schema_comparison = "coluna string, metodo string, aplicavel boolean, outliers long, percentual double"
comparacoes = []
for nome, s, c, *rest in resultados:
    c = c or {}
    for metodo, flag in (("IQR", "iqr_aplicavel"), ("Z_SCORE", "z_aplicavel"), ("MODIFIED_Z_SCORE", "mad_aplicavel")):
        chave = {"IQR":"iqr", "Z_SCORE":"z", "MODIFIED_Z_SCORE":"mad"}[metodo]
        aplicavel = bool(c.get(flag, False))
        qtd = int(c[chave]) if aplicavel else None
        pct = round(100*qtd/s["n"],4) if aplicavel and s["n"] else None
        comparacoes.append((nome, metodo, aplicavel, qtd, pct))
outlier_method_comparison_df = spark.createDataFrame(comparacoes, schema_comparison)
if OUTLIER_DISPLAY_RESULTS:
    display(outlier_method_comparison_df)
````

````python

# ============================================================
# 08.10 RESUMO EXECUTIVO
# ============================================================

# O QUE FAZ:
# Consolida os indicadores gerais de cobertura e as
# sinalizações estatísticas identificadas pelo Outlier Profile.

# COMO FAZ:
# Agrega os resultados do outlier_profile_df utilizando
# funções nativas do PySpark e preserva o schema de saída
# mesmo quando não existem colunas elegíveis.

# POR QUE É IMPORTANTE:
# Garante consistência entre os resultados das etapas,
# evita dependência da estrutura interna de tuplas e
# diferencia sinalizações estatísticas de erros confirmados.

# PERGUNTA RESPONDIDA:
# "Qual é o panorama executivo da detecção de outliers?"

# ------------------------------------------------------------
# 1. VALIDAÇÃO DO DATAFRAME DE ENTRADA
# ------------------------------------------------------------

if "outlier_profile_df" not in globals():
    raise ValueError(
        "Execute a célula 08.8 antes do resumo executivo."
    )

# ------------------------------------------------------------
# 2. AGREGAÇÃO DOS INDICADORES
# ------------------------------------------------------------

resumo_agregado = outlier_profile_df.agg(

    F.count("coluna")
    .cast("long")
    .alias("colunas_avaliadas"),

    F.sum(
        F.when(
            F.col("registros_validos") > 0,
            1
        ).otherwise(0)
    )
    .cast("long")
    .alias("colunas_com_valores_validos"),

    F.sum(
        F.coalesce(
            F.col("outliers_iqr"),
            F.lit(0)
        )
    )
    .cast("long")
    .alias("sinalizacoes_iqr"),

    F.sum(
        F.coalesce(
            F.col("outliers_z"),
            F.lit(0)
        )
    )
    .cast("long")
    .alias("sinalizacoes_z"),

    F.sum(
        F.coalesce(
            F.col("outliers_mad"),
            F.lit(0)
        )
    )
    .cast("long")
    .alias("sinalizacoes_mad"),

    F.sum(
        F.coalesce(
            F.col("outliers_uniao"),
            F.lit(0)
        )
    )
    .cast("long")
    .alias("sinalizacoes_uniao_por_coluna")

)

# ------------------------------------------------------------
# 3. CONSTRUÇÃO DO RESUMO EXECUTIVO
# ------------------------------------------------------------

outlier_summary_df = (
    resumo_agregado

    .select(
        F.lit(int(total_registros))
        .cast("long")
        .alias("total_registros"),

        F.lit(len(df_prepared.columns))
        .cast("int")
        .alias("total_colunas"),

        F.col("colunas_avaliadas"),

        F.coalesce(
            F.col("colunas_com_valores_validos"),
            F.lit(0)
        )
        .cast("long")
        .alias("colunas_com_valores_validos"),

        F.coalesce(
            F.col("sinalizacoes_iqr"),
            F.lit(0)
        )
        .cast("long")
        .alias("sinalizacoes_iqr"),

        F.coalesce(
            F.col("sinalizacoes_z"),
            F.lit(0)
        )
        .cast("long")
        .alias("sinalizacoes_z"),

        F.coalesce(
            F.col("sinalizacoes_mad"),
            F.lit(0)
        )
        .cast("long")
        .alias("sinalizacoes_mad"),

        F.coalesce(
            F.col("sinalizacoes_uniao_por_coluna"),
            F.lit(0)
        )
        .cast("long")
        .alias("sinalizacoes_uniao_por_coluna")
    )
)

# ------------------------------------------------------------
# 4. EXIBIÇÃO DOS RESULTADOS
# ------------------------------------------------------------

if OUTLIER_DISPLAY_RESULTS:
    display(outlier_summary_df)

# ------------------------------------------------------------
# 5. FINALIZAÇÃO
# ------------------------------------------------------------

print("=" * 60)
print("OUTLIER PROFILE - RESUMO EXECUTIVO")
print("=" * 60)

print(
    "Resumo executivo consolidado com sucesso."
)

print(
    "Observação: as somas representam sinalizações "
    "por coluna, não registros únicos do dataset."
)

print(
    "Valores estatisticamente atípicos não representam "
    "necessariamente erros de qualidade."
)
````

````python
# ============================================================
# 08.11 DETALHES OPCIONAIS
# ============================================================

# O QUE FAZ:
# Disponibiliza exemplos limitados de valores sinalizados.

# COMO FAZ:
# Quando habilitado, filtra registros por método e limita os exemplos retornados.

# POR QUE É IMPORTANTE:
# Apoia investigações pontuais sem transferir a base completa para o driver.

# PERGUNTA RESPONDIDA:
# "Quais valores exemplificam as sinalizações estatísticas?"

schema_details = "coluna string, valor string, metodo string"
outlier_details_df = spark.createDataFrame([], schema_details)
if OUTLIER_SHOW_DETAILS:
    exemplos = []
    for nome, s, c in resultados:
        if not c or not c.get("iqr_aplicavel"):
            continue
        x = dq_outlier_col(nome).cast("double")
        flag = (x < F.lit(s["limite_inferior"])) | (x > F.lit(s["limite_superior"]))
        amostra = (df_prepared.where(dq_outlier_valid(dq_outlier_col(nome)) & flag)
                   .select(F.lit(nome).alias("coluna"), dq_outlier_col(nome).cast("string").alias("valor"),
                           F.lit("IQR").alias("metodo"))
                   .limit(OUTLIER_DETAILS_LIMIT))
        exemplos.append(amostra)
    if exemplos:
        from functools import reduce
        outlier_details_df = reduce(lambda a,b: a.unionByName(b), exemplos)
    if OUTLIER_DISPLAY_RESULTS:
        display(outlier_details_df)
````

````python
# ============================================================
# 08.12 TESTES SINTÉTICOS E VALIDAÇÃO DOS RESULTADOS
# ============================================================

# O QUE FAZ:
# Valida os cálculos matemáticos, o tratamento de valores
# especiais e a consistência dos resultados de outliers.

# COMO FAZ:
# Executa testes com dados sintéticos e verifica os contratos
# das estruturas produzidas nas etapas anteriores.

# POR QUE É IMPORTANTE:
# Detecta regressões, inconsistências nas contagens e erros
# de integração entre os componentes do Outlier Profile.

# PERGUNTA RESPONDIDA:
# "Os métodos estatísticos e os resultados consolidados
# apresentam comportamento consistente?"

# ------------------------------------------------------------
# 1. VALIDAÇÃO DAS DEPENDÊNCIAS
# ------------------------------------------------------------

if "resultados" not in globals():
    raise ValueError(
        "Execute a célula 08.7 antes dos testes."
    )

if "estatisticas" not in globals():
    raise ValueError(
        "Execute a célula 08.5 antes dos testes."
    )

# ------------------------------------------------------------
# 2. VALIDAÇÃO DO CONTRATO DE RESULTADOS
# ------------------------------------------------------------

assert all(
    len(resultado) == 4
    for resultado in resultados
), "Cada resultado deve conter quatro componentes."

assert all(
    isinstance(aplicabilidade, dict)
    and set(aplicabilidade) == {"iqr", "z", "mad"}
    for _, _, aplicabilidade, _ in resultados
), "Estrutura de aplicabilidade inválida."

# ------------------------------------------------------------
# 3. TESTES MATEMÁTICOS
# ------------------------------------------------------------

assert dq_outlier_iqr_bounds(
    2.0, 6.0, 1.5
) == (-4.0, 12.0, 4.0)

assert dq_outlier_modified_z_threshold(
    0.0
) is None

assert abs(
    dq_outlier_modified_z_threshold(2.0, 3.5)
    - (3.5 * 2.0 / 0.6745)
) < 1e-10

# ------------------------------------------------------------
# 4. VALIDAÇÃO DAS ESTATÍSTICAS
# ------------------------------------------------------------

assert all(
    0 <= int(s["n"]) <= int(total_registros)
    for s in estatisticas.values()
), "Quantidade de registros válidos inconsistente."

# ------------------------------------------------------------
# 5. VALIDAÇÃO DAS CONTAGENS
# ------------------------------------------------------------

for nome, s, aplicabilidade, contagens in resultados:

    n = int(s["n"])

    if contagens is None:
        assert not any(aplicabilidade.values()), (
            f"{nome}: contagens ausentes para método aplicável."
        )
        continue

    assert any(aplicabilidade.values()), (
        f"{nome}: contagens presentes sem método aplicável."
    )

    for metodo in ("iqr", "z", "mad"):

        quantidade = int(contagens[metodo])

        assert 0 <= quantidade <= n, (
            f"{nome}: contagem inválida para {metodo}."
        )

        if not aplicabilidade[metodo]:
            assert quantidade == 0, (
                f"{nome}: método {metodo} não aplicável "
                "apresentou sinalizações."
            )

    uniao = int(contagens["uniao"])
    intersecao = int(contagens["intersecao_tres"])

    assert 0 <= uniao <= n, (
        f"{nome}: união fora dos limites."
    )

    assert 0 <= intersecao <= n, (
        f"{nome}: interseção fora dos limites."
    )

    assert intersecao <= min(
        int(contagens["iqr"]),
        int(contagens["z"]),
        int(contagens["mad"])
    ), f"{nome}: interseção inconsistente."

    assert uniao >= max(
        int(contagens["iqr"]),
        int(contagens["z"]),
        int(contagens["mad"])
    ), f"{nome}: união inconsistente."

    assert uniao <= sum(
        int(contagens[metodo])
        for metodo in ("iqr", "z", "mad")
    ), f"{nome}: união maior que a soma das sinalizações."

# ------------------------------------------------------------
# 6. TESTE SINTÉTICO DE VALORES ESPECIAIS
# ------------------------------------------------------------

teste_df = spark.createDataFrame(
    [
        (1.0,),
        (1.0,),
        (1.0,),
        (1.0,),
        (100.0,),
        (None,),
        (float("nan"),),
        (float("inf"),),
        (float("-inf"),)
    ],
    "valor DOUBLE"
)

teste_valido = teste_df.where(
    dq_outlier_valid(F.col("valor"))
)

teste_stats = teste_valido.agg(
    F.count("valor").alias("n"),
    F.min("valor").alias("minimo"),
    F.max("valor").alias("maximo")
).first()

assert teste_stats["n"] == 5
assert teste_stats["minimo"] == 1.0
assert teste_stats["maximo"] == 100.0

# ------------------------------------------------------------
# 7. FINALIZAÇÃO
# ------------------------------------------------------------

print("=" * 60)
print("OUTLIER PROFILE - VALIDAÇÃO")
print("=" * 60)

print(f"Colunas verificadas: {len(resultados)}")
print("Contrato de resultados: OK")
print("Cálculos matemáticos: OK")
print("Limites das contagens: OK")
print("Tratamento de valores especiais: OK")
print("Testes concluídos com sucesso.")
````
